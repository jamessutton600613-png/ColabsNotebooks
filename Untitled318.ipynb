````python
# GC "balloon foam" model (2D cross-section)
# - Fixed circular boundary (stiff balloon)
# - Cell division clock + apoptosis
# - Soft repulsion mechanics to resolve overlaps
# - Foam faces visualized via Voronoi polygons clipped to circle
#
# Outputs:
#  - /content/gc_balloon_foam_montage.png
#  - /content/gc_balloon_pressure.png
#  - optional: /content/gc_balloon_foam.gif

import sys, os, math, numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle

# --- dependencies (Colab usually has scipy; shapely may need install)
def ensure_pkg(pkg):
    try:
        __import__(pkg)
        return True
    except Exception:
        return False

if not ensure_pkg("scipy"):
    !pip -q install scipy
if not ensure_pkg("shapely"):
    !pip -q install shapely

from scipy.spatial import Voronoi
from shapely.geometry import Polygon, Point
from shapely.ops import unary_union

rng = np.random.default_rng(7)

# =========================
# Parameters
# =========================
R = 1.0                      # balloon radius (fixed)
N_init = 55                  # initial cells
N_cap = 220                  # hard cap (prevents blowup)
steps = 420                  # simulation steps
dt = 0.03                    # mechanics step
damp = 0.90                  # velocity damping

r_cell = 0.075               # nominal cell radius (size constrained by balloon)
k_rep = 55.0                 # repulsion stiffness
k_wall = 120.0               # wall stiffness

# Division / death kinetics (dimensionless step rates)
# Think "division period ~ 8-12 hours" as step counts in [120, 180]
div_min = 120
div_max = 180

death_rate = 0.010           # per-step probability per cell (tune)
death_pressure_gain = 0.012  # increases death with pressure (homeostasis-like)

birth_enabled = True         # if False, no division
death_enabled = True         # if False, no apoptosis

# Snapshot settings
snapshots = 12               # montage panels
make_gif = False             # set True if you want a gif (slower)

# =========================
# Helpers
# =========================
def rand_in_disc(R_inner):
    u = rng.random()
    ang = 2*np.pi*rng.random()
    rr = R_inner * np.sqrt(u)
    return np.array([rr*np.cos(ang), rr*np.sin(ang)], dtype=float)

def clip_to_balloon(x):
    n = np.linalg.norm(x)
    if n <= R - r_cell:
        return x
    return x * ((R - r_cell) / (n + 1e-12))

def balloon_force(x):
    # hard-ish wall: if center goes beyond (R - r_cell), push inward
    n = np.linalg.norm(x) + 1e-12
    limit = R - r_cell
    if n <= limit:
        return np.zeros(2)
    # radial inward force proportional to penetration
    pen = n - limit
    return -k_wall * pen * (x / n)

def pressure_proxy(overlap):
    # mean overlap is a decent "pressure" proxy here
    return float(np.mean(overlap))

def circle_polygon(R, n=256):
    pts = [(R*math.cos(2*math.pi*i/n), R*math.sin(2*math.pi*i/n)) for i in range(n)]
    return Polygon(pts)

BALL = circle_polygon(R, n=256)

# Voronoi helpers (finite polygons)
def voronoi_finite_polygons_2d(vor, radius=5.0):
    """
    Reconstruct infinite Voronoi regions to finite regions.
    Returns (regions, vertices). Standard recipe.
    """
    new_regions = []
    new_vertices = vor.vertices.tolist()

    center = vor.points.mean(axis=0)
    if vor.points.shape[1] != 2:
        raise ValueError("Requires 2D input")

    # Map ridge vertices to ridges
    all_ridges = {}
    for (p1, p2), (v1, v2) in zip(vor.ridge_points, vor.ridge_vertices):
        all_ridges.setdefault(p1, []).append((p2, v1, v2))
        all_ridges.setdefault(p2, []).append((p1, v1, v2))

    for p1, region_idx in enumerate(vor.point_region):
        vertices = vor.regions[region_idx]
        if all(v >= 0 for v in vertices):
            new_regions.append(vertices)
            continue

        # reconstruct region
        ridges = all_ridges[p1]
        new_region = [v for v in vertices if v >= 0]

        for p2, v1, v2 in ridges:
            if v2 < 0 or v1 < 0:
                v = v1 if v1 >= 0 else v2
                # direction
                t = vor.points[p2] - vor.points[p1]
                t = t / (np.linalg.norm(t) + 1e-12)
                n = np.array([-t[1], t[0]])  # normal
                midpoint = (vor.points[p1] + vor.points[p2]) / 2
                direction = np.sign(np.dot(midpoint - center, n)) * n
                far_point = vor.vertices[v] + direction * radius
                new_vertices.append(far_point.tolist())
                new_region.append(len(new_vertices) - 1)

        # sort vertices CCW
        vs = np.asarray([new_vertices[v] for v in new_region])
        c = vs.mean(axis=0)
        angles = np.arctan2(vs[:,1] - c[1], vs[:,0] - c[0])
        new_region = [v for _, v in sorted(zip(angles, new_region))]
        new_regions.append(new_region)

    return new_regions, np.asarray(new_vertices)

def clipped_voronoi_polys(points):
    """
    Returns list of shapely Polygons (one per point), clipped to the balloon circle.
    """
    if len(points) < 4:
        # Voronoi needs >=4 points; fallback to tiny circles as polygons
        polys = []
        for p in points:
            polys.append(Point(p[0], p[1]).buffer(r_cell*1.05))
        return polys

    vor = Voronoi(points)
    regions, vertices = voronoi_finite_polygons_2d(vor, radius=4.0)

    polys = []
    for reg in regions:
        poly = Polygon(vertices[reg])
        poly = poly.intersection(BALL)
        if not poly.is_empty and poly.area > 1e-6:
            polys.append(poly)
        else:
            # fallback
            polys.append(Point(vor.points[len(polys)][0], vor.points[len(polys)][1]).buffer(r_cell))
    return polys

# =========================
# State
# =========================
pos = np.stack([rand_in_disc(R*0.75) for _ in range(N_init)], axis=0)
vel = rng.normal(scale=0.01, size=pos.shape)

# Division clocks
div_clock = rng.integers(div_min, div_max+1, size=N_init)

# Clone IDs (coloring only)
n_start_clones = 8
clone_id = rng.integers(0, n_start_clones, size=N_init)
next_clone = int(np.max(clone_id)) + 1

# =========================
# Dynamics
# =========================
def mechanics_step(pos, vel):
    N = len(pos)
    # pairwise
    dx = pos[:,None,:] - pos[None,:,:]
    dist = np.linalg.norm(dx, axis=2) + 1e-12
    np.fill_diagonal(dist, 1e9)

    # overlap
    target = 2.0 * r_cell
    overlap = np.maximum(target - dist, 0.0)

    ux = dx[:,:,0] / dist
    uy = dx[:,:,1] / dist

    fx = k_rep * np.sum(overlap * ux, axis=1)
    fy = k_rep * np.sum(overlap * uy, axis=1)
    F = np.stack([fx, fy], axis=1)

    # wall forces
    for i in range(N):
        F[i] += balloon_force(pos[i])

    vel = damp*vel + dt*F
    pos = pos + dt*vel

    # clamp to balloon interior
    for i in range(N):
        pos[i] = clip_to_balloon(pos[i])

    return pos, vel, overlap

def division_step():
    global pos, vel, div_clock, clone_id, next_clone
    if not birth_enabled:
        return

    N = len(pos)
    if N >= N_cap:
        return

    # decrement clocks
    div_clock -= 1

    # which divide?
    to_div = np.where(div_clock <= 0)[0]
    if to_div.size == 0:
        return

    # divide a limited number per step to avoid explosion
    rng.shuffle(to_div)
    to_div = to_div[:max(1, int(0.08 * N))]

    for p in to_div:
        if len(pos) >= N_cap:
            break

        # create daughter near parent
        jitter = rng.normal(scale=r_cell*0.45, size=2)
        x_new = clip_to_balloon(pos[p] + jitter)
        v_new = vel[p] + rng.normal(scale=0.01, size=2)

        pos = np.vstack([pos, x_new[None,:]])
        vel = np.vstack([vel, v_new[None,:]])

        # reset parent clock and set daughter clock
        div_clock[p] = int(rng.integers(div_min, div_max+1))
        div_clock = np.append(div_clock, int(rng.integers(div_min, div_max+1)))

        # clone inheritance (mostly same clone; occasional new clone)
        if rng.random() < 0.03 and next_clone < 40:
            c_new = next_clone
            next_clone += 1
        else:
            c_new = int(clone_id[p])
        clone_id = np.append(clone_id, c_new)

def death_step(pressure_scalar):
    global pos, vel, div_clock, clone_id
    if not death_enabled:
        return

    N = len(pos)
    if N <= 12:
        return

    # increase death when pressure high (homeostasis)
    p = float(death_rate + death_pressure_gain * max(pressure_scalar, 0.0))
    p = min(p, 0.08)  # cap

    die = rng.random(N) < p
    # keep at least 12
    if np.sum(~die) < 12:
        return

    keep = ~die
    pos = pos[keep]
    vel = vel[keep]
    div_clock = div_clock[keep]
    clone_id = clone_id[keep]

def dominant_clone_fraction(clone_id):
    if len(clone_id) == 0:
        return 0.0
    _, counts = np.unique(clone_id, return_counts=True)
    return float(np.max(counts) / (np.sum(counts) + 1e-12))

# =========================
# Run + collect snapshots
# =========================
snap_idx = np.linspace(0, steps-1, snapshots, dtype=int).tolist()
snaps = []
pressure_series = []
N_series = []
dom_series = []

for t in range(steps):
    pos, vel, overlap = mechanics_step(pos, vel)
    psc = pressure_proxy(overlap)
    pressure_series.append(psc)
    N_series.append(len(pos))
    dom_series.append(dominant_clone_fraction(clone_id))

    death_step(psc)
    division_step()

    if t in snap_idx:
        snaps.append(dict(t=t, pos=pos.copy(), clone=clone_id.copy()))

# =========================
# Plot montage with Voronoi foam faces
# =========================
rows, cols = 3, 4
fig, axes = plt.subplots(rows, cols, figsize=(10.8, 8.4), dpi=170, constrained_layout=True)
cmap = plt.get_cmap("tab20")

for ax, fr in zip(axes.ravel(), snaps):
    ax.set_aspect("equal", "box")
    ax.set_xlim(-R, R)
    ax.set_ylim(-R, R)
    ax.axis("off")

    # balloon boundary
    ax.add_patch(Circle((0,0), R, fill=False, linewidth=0.8, edgecolor="black"))

    P = fr["pos"]
    C = fr["clone"]

    polys = clipped_voronoi_polys(P)

    # draw cells as foam faces (polygons)
    for i, poly in enumerate(polys):
        if poly.is_empty:
            continue
        x, y = poly.exterior.xy
        col = cmap(int(C[i]) % 20)
        ax.fill(x, y, color=col, alpha=0.55, linewidth=0.3, edgecolor="black")

    # centers for reference (small)
    ax.scatter(P[:,0], P[:,1], s=3, c="black", alpha=0.55)

    dom = dominant_clone_fraction(C)
    ax.set_title(f"t={fr['t']:03d} | N={len(P)} | dom={dom:.2f}", fontsize=8)

for ax in axes.ravel()[len(snaps):]:
    ax.axis("off")

montage_path = "/content/gc_balloon_foam_montage.png"
plt.savefig(montage_path, bbox_inches="tight")
plt.show()

# =========================
# Plot pressure vs time
# =========================
plt.figure(figsize=(8.2, 3.2), dpi=170)
plt.plot(pressure_series, linewidth=1.2)
plt.xlabel("time step")
plt.ylabel("mean overlap (pressure proxy)")
plt.title("GC balloon packing pressure over time")
plt.tight_layout()
pressure_path = "/content/gc_balloon_pressure.png"
plt.savefig(pressure_path, bbox_inches="tight")
plt.show()

print("Saved montage:", montage_path)
print("Saved pressure plot:", pressure_path)
````

````python
# Advanced GC Balloon-Foam (2D cross-section)
# Physics: confined active foam + local pressure + delayed apoptosis + stress-accelerated clearance
# CS: contact graph + effective dimensionality + variance/ridge indicators
# Immunogenetics: clone manifolds + face-length weighted help

import math, os
import numpy as np
import matplotlib.pyplot as plt

# deps
def ensure(pkg):
    try:
        __import__(pkg)
        return True
    except Exception:
        return False

if not ensure("scipy"):
    !pip -q install scipy
if not ensure("shapely"):
    !pip -q install shapely

from scipy.spatial import Voronoi
from shapely.geometry import Polygon, Point
from shapely.ops import unary_union

rng = np.random.default_rng(11)

# -----------------------------
# Core parameters (tune here)
# -----------------------------
R = 1.0                       # GC balloon radius
r_cell = 0.070                # nominal radius (size constraint)
steps = 520                   # simulation steps
dt = 0.03
damp = 0.90

N_init = 60
N_cap = 240                   # safety cap (still useful to prevent blowups)

# mechanics
k_rep = 70.0                  # repulsion stiffness (overlap force)
k_wall = 150.0                # boundary stiffness

# division clocks (steps)
div_min = 120                 # "fastest division"
div_max = 180

# apoptosis model (delayed)
# doom_timer: "pre-apoptotic" phase (still present but weakening)
# exec_timer: "execution" phase (rapid shrink; then removed)
doom_min = 160                # ~1.0x fastest division (your requested scale)
doom_max = 240                # 1.3–1.5x fastest division
exec_min = 20                 # fast execution
exec_max = 40

# pressure accelerates apoptosis progression (foam crush)
pressure_accel = 1.2          # 0 = none; 1–2 = moderate

# base hazard (alive -> doomed), pressure-dependent
death_base = 0.004
death_pressure_gain = 0.014

# face-weighted help (antigen/Tfh proxy)
help_gain = 0.005             # gain per step per unit face-length contact to "FDC zone"
help_decay = 0.92             # help accumulator decay
help_threshold = 0.35         # division boost if above (acts like LZ licensing)
help_boost = 0.55             # boosts division probability/clock speed if helped

# "FDC antigen depot" proxy: a fixed region (light-zone-ish)
FDC_center = np.array([0.55, 0.0])
FDC_radius = 0.35

# clones / manifolds
n_start_clones = 10
new_clone_prob = 0.03
max_clones = 60

# snapshots
snapshots = 12

# -----------------------------
# Geometry helpers
# -----------------------------
def circle_polygon(R, n=256):
    pts = [(R*math.cos(2*math.pi*i/n), R*math.sin(2*math.pi*i/n)) for i in range(n)]
    return Polygon(pts)

BALL = circle_polygon(R, 256)

def rand_in_disc(R_inner):
    u = rng.random()
    ang = 2*np.pi*rng.random()
    rr = R_inner * np.sqrt(u)
    return np.array([rr*np.cos(ang), rr*np.sin(ang)], dtype=float)

def clip_to_balloon(x):
    n = np.linalg.norm(x)
    lim = R - r_cell
    if n <= lim:
        return x
    return x * (lim / (n + 1e-12))

def wall_force(x):
    n = np.linalg.norm(x) + 1e-12
    lim = R - r_cell
    if n <= lim:
        return np.zeros(2)
    pen = n - lim
    return -k_wall * pen * (x / n)

# finite Voronoi reconstruction
def voronoi_finite_polygons_2d(vor, radius=5.0):
    new_regions = []
    new_vertices = vor.vertices.tolist()
    center = vor.points.mean(axis=0)

    all_ridges = {}
    for (p1, p2), (v1, v2) in zip(vor.ridge_points, vor.ridge_vertices):
        all_ridges.setdefault(p1, []).append((p2, v1, v2))
        all_ridges.setdefault(p2, []).append((p1, v1, v2))

    for p1, reg_idx in enumerate(vor.point_region):
        verts = vor.regions[reg_idx]
        if all(v >= 0 for v in verts):
            new_regions.append(verts)
            continue

        ridges = all_ridges[p1]
        new_region = [v for v in verts if v >= 0]

        for p2, v1, v2 in ridges:
            if v1 < 0 or v2 < 0:
                v = v1 if v1 >= 0 else v2
                t = vor.points[p2] - vor.points[p1]
                t = t / (np.linalg.norm(t) + 1e-12)
                n = np.array([-t[1], t[0]])
                midpoint = (vor.points[p1] + vor.points[p2]) / 2
                direction = np.sign(np.dot(midpoint - center, n)) * n
                far_point = vor.vertices[v] + direction * radius
                new_vertices.append(far_point.tolist())
                new_region.append(len(new_vertices) - 1)

        vs = np.asarray([new_vertices[v] for v in new_region])
        c = vs.mean(axis=0)
        angles = np.arctan2(vs[:,1] - c[1], vs[:,0] - c[0])
        new_region = [v for _, v in sorted(zip(angles, new_region))]
        new_regions.append(new_region)

    return new_regions, np.asarray(new_vertices)

def clipped_voronoi(points):
    if len(points) < 4:
        polys = []
        for p in points:
            polys.append(Point(p[0], p[1]).buffer(r_cell*1.02).intersection(BALL))
        return polys

    vor = Voronoi(points)
    regions, vertices = voronoi_finite_polygons_2d(vor, radius=4.0)
    polys = []
    for reg in regions:
        poly = Polygon(vertices[reg]).intersection(BALL)
        if poly.is_empty or poly.area < 1e-6:
            poly = Point(vor.points[len(polys)][0], vor.points[len(polys)][1]).buffer(r_cell).intersection(BALL)
        polys.append(poly)
    return polys

def simpson_Neff(clone_ids):
    if len(clone_ids) == 0:
        return 0.0
    _, counts = np.unique(clone_ids, return_counts=True)
    f = counts / (np.sum(counts) + 1e-12)
    return float(1.0 / (np.sum(f*f) + 1e-12))

def dom_fraction(clone_ids):
    if len(clone_ids) == 0:
        return 0.0
    _, counts = np.unique(clone_ids, return_counts=True)
    return float(np.max(counts) / (np.sum(counts) + 1e-12))

def in_FDC_zone(x):
    return np.linalg.norm(x - FDC_center) <= FDC_radius

# -----------------------------
# State
# -----------------------------
pos = np.stack([rand_in_disc(R*0.75) for _ in range(N_init)], axis=0)
vel = rng.normal(scale=0.01, size=pos.shape)

clone_id = rng.integers(0, n_start_clones, size=N_init)
next_clone = int(np.max(clone_id)) + 1

# help accumulator per cell (proxy for LZ licensing)
help_acc = np.zeros(N_init, dtype=float)

# division clocks
div_clock = rng.integers(div_min, div_max+1, size=N_init)

# apoptosis state
# 0 = alive, 1 = doomed, 2 = executing
state = np.zeros(N_init, dtype=np.int8)
doom_timer = np.zeros(N_init, dtype=float)
exec_timer = np.zeros(N_init, dtype=float)

# -----------------------------
# Mechanics / pressure
# -----------------------------
def mechanics_step(pos, vel):
    N = len(pos)
    dx = pos[:,None,:] - pos[None,:,:]
    dist = np.linalg.norm(dx, axis=2) + 1e-12
    np.fill_diagonal(dist, 1e9)

    target = 2.0 * r_cell
    overlap = np.maximum(target - dist, 0.0)

    ux = dx[:,:,0] / dist
    uy = dx[:,:,1] / dist
    fx = k_rep * np.sum(overlap * ux, axis=1)
    fy = k_rep * np.sum(overlap * uy, axis=1)
    F = np.stack([fx, fy], axis=1)

    for i in range(N):
        F[i] += wall_force(pos[i])

    vel = damp*vel + dt*F
    pos = pos + dt*vel
    for i in range(N):
        pos[i] = clip_to_balloon(pos[i])

    # per-cell pressure proxy: sum overlaps
    P_i = np.sum(overlap, axis=1)
    P_mean = float(np.mean(P_i)) if N > 0 else 0.0
    return pos, vel, P_i, P_mean

# -----------------------------
# Face-length weighted help
# -----------------------------
def face_weighted_help(polys, pos):
    """
    Approx: reward a cell if its Voronoi polygon overlaps the fixed FDC region.
    Use perimeter length of intersection boundary as proxy for "face contact capacity".
    """
    fdc = Point(FDC_center[0], FDC_center[1]).buffer(FDC_radius)
    gains = np.zeros(len(polys), dtype=float)
    for i, poly in enumerate(polys):
        inter = poly.intersection(fdc)
        if inter.is_empty:
            continue
        # boundary length of intersection region approximates interface opportunity
        gains[i] = float(inter.length)
    return gains

# -----------------------------
# Apoptosis transitions
# -----------------------------
def mark_doomed(i):
    state[i] = 1
    doom_timer[i] = float(rng.integers(doom_min, doom_max+1))
    exec_timer[i] = float(rng.integers(exec_min, exec_max+1))

def death_hazard(P_i, P_mean, help_acc):
    # hazard for alive -> doomed
    # pressure raises hazard; help lowers hazard
    Pn = P_i / (P_mean + 1e-12)
    p = death_base + death_pressure_gain * max(Pn - 1.0, 0.0)
    p = p * (1.0 - 0.55*np.tanh(help_acc))  # help protects
    return float(np.clip(p, 0.0, 0.08))

def advance_apoptosis(P_i, P_mean):
    """
    Decrement doom/exec timers; pressure accelerates.
    Executing cells shrink in effective influence (modeled by reduced repulsion via smaller r_cell factor—approx).
    Here we implement shrink by pushing them inward slightly + increasing death progression.
    """
    global state, doom_timer, exec_timer
    N = len(state)
    if N == 0:
        return

    accel = 1.0 + pressure_accel * (P_i / (P_mean + 1e-12))
    accel = np.clip(accel, 0.5, 6.0)

    # doomed timers
    doomed = (state == 1)
    doom_timer[doomed] -= accel[doomed]
    # when doom ends -> executing
    to_exec = doomed & (doom_timer <= 0)
    state[to_exec] = 2

    # executing timers
    executing = (state == 2)
    exec_timer[executing] -= accel[executing]

def remove_completed():
    """
    Remove executing cells whose exec_timer <= 0
    """
    global pos, vel, clone_id, help_acc, div_clock, state, doom_timer, exec_timer
    done = (state == 2) & (exec_timer <= 0)
    if np.any(done):
        keep = ~done
        pos = pos[keep]
        vel = vel[keep]
        clone_id = clone_id[keep]
        help_acc = help_acc[keep]
        div_clock = div_clock[keep]
        state = state[keep]
        doom_timer = doom_timer[keep]
        exec_timer = exec_timer[keep]

# -----------------------------
# Division (help-sensitive)
# -----------------------------
def division_step():
    global pos, vel, clone_id, next_clone, help_acc, div_clock, state, doom_timer, exec_timer
    if len(pos) >= N_cap:
        return

    N = len(pos)
    # only alive or doomed can divide? (executing cannot)
    can_div = (state != 2)

    # decrement clocks, faster if helped (LZ licensing)
    speed = 1.0 + help_boost * (help_acc > help_threshold)
    div_clock[can_div] -= speed[can_div].astype(int)

    to_div = np.where((div_clock <= 0) & can_div)[0]
    if to_div.size == 0:
        return

    rng.shuffle(to_div)
    to_div = to_div[:max(1, int(0.08*N))]

    for p in to_div:
        if len(pos) >= N_cap:
            break

        # daughter near parent
        jitter = rng.normal(scale=r_cell*0.45, size=2)
        x_new = clip_to_balloon(pos[p] + jitter)
        v_new = vel[p] + rng.normal(scale=0.01, size=2)

        pos = np.vstack([pos, x_new[None,:]])
        vel = np.vstack([vel, v_new[None,:]])

        # clone inheritance; occasional new clone
        if rng.random() < new_clone_prob and next_clone < max_clones:
            c_new = next_clone
            next_clone += 1
        else:
            c_new = int(clone_id[p])
        clone_id = np.append(clone_id, c_new)

        # new cell arrays
        help_acc = np.append(help_acc, 0.0)
        div_clock[p] = int(rng.integers(div_min, div_max+1))
        div_clock = np.append(div_clock, int(rng.integers(div_min, div_max+1)))

        # apoptosis state for new
        state = np.append(state, 0).astype(np.int8)
        doom_timer = np.append(doom_timer, 0.0)
        exec_timer = np.append(exec_timer, 0.0)

# -----------------------------
# Simulation loop + logging
# -----------------------------
snap_idx = np.linspace(0, steps-1, snapshots, dtype=int).tolist()
snaps = []

P_series = []
N_series = []
dom_series = []
Neff_series = []
var_series = []   # instability proxy: rolling variance of pressure

for t in range(steps):
    # build Voronoi foam
    polys = clipped_voronoi(pos)

    # face-weighted help (FDC zone)
    gain = face_weighted_help(polys, pos)
    help_acc = help_decay*help_acc + help_gain*gain

    # mechanics / pressure
    pos, vel, P_i, P_mean = mechanics_step(pos, vel)
    P_series.append(P_mean)
    N_series.append(len(pos))
    dom_series.append(dom_fraction(clone_id))
    Neff_series.append(simpson_Neff(clone_id))

    # alive -> doomed events
    alive = np.where(state == 0)[0]
    for i in alive:
        p = death_hazard(P_i[i], P_mean, help_acc[i])
        if rng.random() < p:
            mark_doomed(i)

    # advance delayed apoptosis and remove completed
    advance_apoptosis(P_i, P_mean)
    remove_completed()

    # division (help-sensitive)
    division_step()

    # log instability proxy: rolling variance over last 25 steps
    if len(P_series) >= 25:
        var_series.append(float(np.var(P_series[-25:])))
    else:
        var_series.append(0.0)

    if t in snap_idx:
        snaps.append(dict(t=t, pos=pos.copy(), clone=clone_id.copy(), state=state.copy()))

# -----------------------------
# Plots
# -----------------------------
# 1) Foam montage
rows, cols = 3, 4
fig, axes = plt.subplots(rows, cols, figsize=(10.8, 8.4), dpi=170, constrained_layout=True)
cmap = plt.get_cmap("tab20")

fdc_circle = Point(FDC_center[0], FDC_center[1]).buffer(FDC_radius)

for ax, fr in zip(axes.ravel(), snaps):
    ax.set_aspect("equal", "box")
    ax.set_xlim(-R, R)
    ax.set_ylim(-R, R)
    ax.axis("off")

    ax.add_patch(plt.Circle((0,0), R, fill=False, linewidth=0.8, edgecolor="black"))

    # show FDC region
    fx, fy = fdc_circle.exterior.xy
    ax.plot(fx, fy, linewidth=0.6)

    P = fr["pos"]
    C = fr["clone"]
    S = fr["state"]

    polys = clipped_voronoi(P)

    for i, poly in enumerate(polys):
        if poly.is_empty:
            continue
        x, y = poly.exterior.xy
        col = cmap(int(C[i]) % 20)

        alpha = 0.55
        if S[i] == 1:   # doomed
            alpha = 0.35
        if S[i] == 2:   # executing
            alpha = 0.18

        ax.fill(x, y, color=col, alpha=alpha, linewidth=0.25, edgecolor="black")

    ax.scatter(P[:,0], P[:,1], s=2, c="black", alpha=0.50)

    ax.set_title(f"t={fr['t']:03d} | N={len(P)} | dom={dom_fraction(C):.2f} | Neff={simpson_Neff(C):.1f}", fontsize=8)

for ax in axes.ravel()[len(snaps):]:
    ax.axis("off")

montage_path = "/content/gc_balloon_foam_advanced_montage.png"
plt.savefig(montage_path, bbox_inches="tight")
plt.show()

# 2) Timeseries
t = np.arange(len(P_series))

plt.figure(figsize=(10.4, 3.2), dpi=170)
plt.plot(t, P_series, linewidth=1.2)
plt.title("Pressure proxy (mean overlap)")
plt.xlabel("time step")
plt.ylabel("pressure")
plt.tight_layout()
pressure_path = "/content/gc_balloon_pressure_advanced.png"
plt.savefig(pressure_path, bbox_inches="tight")
plt.show()

plt.figure(figsize=(10.4, 3.2), dpi=170)
plt.plot(t, Neff_series, linewidth=1.2)
plt.title(r"Effective manifold number $N_{\mathrm{eff}}$ (Simpson)")
plt.xlabel("time step")
plt.ylabel("Neff")
plt.tight_layout()
neff_path = "/content/gc_balloon_neff.png"
plt.savefig(neff_path, bbox_inches="tight")
plt.show()

plt.figure(figsize=(10.4, 3.2), dpi=170)
plt.plot(t, dom_series, linewidth=1.2)
plt.title("Dominant clone fraction")
plt.xlabel("time step")
plt.ylabel("dominant fraction")
plt.tight_layout()
dom_path = "/content/gc_balloon_dominance.png"
plt.savefig(dom_path, bbox_inches="tight")
plt.show()

plt.figure(figsize=(10.4, 3.2), dpi=170)
plt.plot(t, var_series, linewidth=1.2)
plt.title("Instability proxy: rolling variance of pressure (ridge indicator)")
plt.xlabel("time step")
plt.ylabel("Var(P) over last 25 steps")
plt.tight_layout()
var_path = "/content/gc_balloon_variance.png"
plt.savefig(var_path, bbox_inches="tight")
plt.show()

print("Saved:", montage_path)
print("Saved:", pressure_path)
print("Saved:", neff_path)
print("Saved:", dom_path)
print("Saved:", var_path)
````

````python
# ============================================================
# Advanced GC Balloon-Foam + Macro-Jump + Lineage Honeycomb Plot
# Full standalone Colab block (no line numbers)
# ============================================================

import math, os
import numpy as np
import matplotlib.pyplot as plt

# deps
def ensure(pkg):
    try:
        __import__(pkg)
        return True
    except Exception:
        return False

if not ensure("scipy"):
    !pip -q install scipy
if not ensure("shapely"):
    !pip -q install shapely

from scipy.spatial import Voronoi
from shapely.geometry import Polygon, Point

rng = np.random.default_rng(11)

# -----------------------------
# Parameters (tune here)
# -----------------------------
R = 1.0
r_cell = 0.070
steps = 520
dt = 0.03
damp = 0.90

N_init = 60
N_cap = 260  # safety cap

k_rep = 70.0
k_wall = 150.0

div_min = 120
div_max = 180

# delayed apoptosis
doom_min = 160
doom_max = 240
exec_min = 20
exec_max = 40

pressure_accel = 1.2
death_base = 0.004
death_pressure_gain = 0.014

# face-weighted help
help_gain = 0.005
help_decay = 0.92
help_threshold = 0.35
help_boost = 0.55

FDC_center = np.array([0.55, 0.0])
FDC_radius = 0.35

# clones / manifolds
n_start_clones = 10
max_clones = 120

# Macro-jump (internal new manifold creation)
jump_base = 0.002        # baseline macro-jump chance at division
jump_help_gain = 0.035   # increases jump when help is low
jump_press_gain = 0.040  # increases jump when pressure is high
jump_sigmoid_k = 6.0     # steepness of gating

# snapshots
snapshots = 12

# ------------------------------------------------------------
# Helpers
# ------------------------------------------------------------
def sigmoid(z):
    z = np.clip(z, -60.0, 60.0)
    return 1.0 / (1.0 + np.exp(-z))

def circle_polygon(R, n=256):
    pts = [(R*math.cos(2*math.pi*i/n), R*math.sin(2*math.pi*i/n)) for i in range(n)]
    return Polygon(pts)

BALL = circle_polygon(R, 256)
FDC = Point(FDC_center[0], FDC_center[1]).buffer(FDC_radius)

def rand_in_disc(R_inner):
    u = rng.random()
    ang = 2*np.pi*rng.random()
    rr = R_inner * np.sqrt(u)
    return np.array([rr*np.cos(ang), rr*np.sin(ang)], dtype=float)

def clip_to_balloon(x):
    n = np.linalg.norm(x)
    lim = R - r_cell
    if n <= lim:
        return x
    return x * (lim / (n + 1e-12))

def wall_force(x):
    n = np.linalg.norm(x) + 1e-12
    lim = R - r_cell
    if n <= lim:
        return np.zeros(2)
    pen = n - lim
    return -k_wall * pen * (x / n)

# Finite Voronoi reconstruction
def voronoi_finite_polygons_2d(vor, radius=5.0):
    new_regions = []
    new_vertices = vor.vertices.tolist()
    center = vor.points.mean(axis=0)

    all_ridges = {}
    for (p1, p2), (v1, v2) in zip(vor.ridge_points, vor.ridge_vertices):
        all_ridges.setdefault(p1, []).append((p2, v1, v2))
        all_ridges.setdefault(p2, []).append((p1, v1, v2))

    for p1, reg_idx in enumerate(vor.point_region):
        verts = vor.regions[reg_idx]
        if all(v >= 0 for v in verts):
            new_regions.append(verts)
            continue

        ridges = all_ridges[p1]
        new_region = [v for v in verts if v >= 0]

        for p2, v1, v2 in ridges:
            if v1 < 0 or v2 < 0:
                v = v1 if v1 >= 0 else v2
                t = vor.points[p2] - vor.points[p1]
                t = t / (np.linalg.norm(t) + 1e-12)
                n = np.array([-t[1], t[0]])
                midpoint = (vor.points[p1] + vor.points[p2]) / 2
                direction = np.sign(np.dot(midpoint - center, n)) * n
                far_point = vor.vertices[v] + direction * radius
                new_vertices.append(far_point.tolist())
                new_region.append(len(new_vertices) - 1)

        vs = np.asarray([new_vertices[v] for v in new_region])
        c = vs.mean(axis=0)
        angles = np.arctan2(vs[:,1] - c[1], vs[:,0] - c[0])
        new_region = [v for _, v in sorted(zip(angles, new_region))]
        new_regions.append(new_region)

    return new_regions, np.asarray(new_vertices)

def clipped_voronoi(points):
    if len(points) < 4:
        polys = []
        for p in points:
            polys.append(Point(p[0], p[1]).buffer(r_cell*1.02).intersection(BALL))
        return polys

    vor = Voronoi(points)
    regions, vertices = voronoi_finite_polygons_2d(vor, radius=4.0)
    polys = []
    for reg in regions:
        poly = Polygon(vertices[reg]).intersection(BALL)
        if poly.is_empty or poly.area < 1e-6:
            poly = Point(vor.points[len(polys)][0], vor.points[len(polys)][1]).buffer(r_cell).intersection(BALL)
        polys.append(poly)
    return polys

def simpson_Neff(clone_ids):
    if len(clone_ids) == 0:
        return 0.0
    _, counts = np.unique(clone_ids, return_counts=True)
    f = counts / (np.sum(counts) + 1e-12)
    return float(1.0 / (np.sum(f*f) + 1e-12))

def dom_fraction(clone_ids):
    if len(clone_ids) == 0:
        return 0.0
    _, counts = np.unique(clone_ids, return_counts=True)
    return float(np.max(counts) / (np.sum(counts) + 1e-12))

def face_weighted_help(polys):
    gains = np.zeros(len(polys), dtype=float)
    for i, poly in enumerate(polys):
        inter = poly.intersection(FDC)
        if inter.is_empty:
            continue
        gains[i] = float(inter.length)
    return gains

# -----------------------------
# State arrays
# -----------------------------
pos = np.stack([rand_in_disc(R*0.75) for _ in range(N_init)], axis=0)
vel = rng.normal(scale=0.01, size=pos.shape)

clone_id = rng.integers(0, n_start_clones, size=N_init)
next_clone = int(np.max(clone_id)) + 1

help_acc = np.zeros(N_init, dtype=float)
div_clock = rng.integers(div_min, div_max+1, size=N_init)

# apoptosis state: 0 alive, 1 doomed, 2 executing
state = np.zeros(N_init, dtype=np.int8)
doom_timer = np.zeros(N_init, dtype=float)
exec_timer = np.zeros(N_init, dtype=float)

# genealogy bookkeeping (IDs persist even if removed)
# Each cell has a unique integer "cell_uid"
next_uid = N_init
cell_uid = np.arange(N_init, dtype=int)
parent_uid = np.full(N_init, -1, dtype=int)
birth_t = np.zeros(N_init, dtype=int)
death_t = np.full(N_init, -1, dtype=int)  # set when removed
clone_at_birth = clone_id.copy()

# We keep global lists to reconstruct tree even after deletions
G_parent = {int(u): int(p) for u, p in zip(cell_uid, parent_uid)}
G_birth = {int(u): int(t) for u, t in zip(cell_uid, birth_t)}
G_death = {}  # filled when removed
G_clone = {int(u): int(c) for u, c in zip(cell_uid, clone_at_birth)}

# -----------------------------
# Mechanics step
# -----------------------------
def mechanics_step(pos, vel):
    N = len(pos)
    dx = pos[:,None,:] - pos[None,:,:]
    dist = np.linalg.norm(dx, axis=2) + 1e-12
    np.fill_diagonal(dist, 1e9)

    target = 2.0 * r_cell
    overlap = np.maximum(target - dist, 0.0)

    ux = dx[:,:,0] / dist
    uy = dx[:,:,1] / dist
    fx = k_rep * np.sum(overlap * ux, axis=1)
    fy = k_rep * np.sum(overlap * uy, axis=1)
    F = np.stack([fx, fy], axis=1)

    for i in range(N):
        F[i] += wall_force(pos[i])

    vel = damp*vel + dt*F
    pos = pos + dt*vel
    for i in range(N):
        pos[i] = clip_to_balloon(pos[i])

    P_i = np.sum(overlap, axis=1)
    P_mean = float(np.mean(P_i)) if N > 0 else 0.0
    return pos, vel, P_i, P_mean

# -----------------------------
# Apoptosis functions
# -----------------------------
def mark_doomed(i):
    state[i] = 1
    doom_timer[i] = float(rng.integers(doom_min, doom_max+1))
    exec_timer[i] = float(rng.integers(exec_min, exec_max+1))

def death_hazard(P_i, P_mean, h):
    Pn = float(P_i / (P_mean + 1e-12))
    p = death_base + death_pressure_gain * max(Pn - 1.0, 0.0)
    p = p * (1.0 - 0.55*np.tanh(h))
    return float(np.clip(p, 0.0, 0.08))

def advance_apoptosis(P_i, P_mean):
    accel = 1.0 + pressure_accel * (P_i / (P_mean + 1e-12))
    accel = np.clip(accel, 0.5, 6.0)

    doomed = (state == 1)
    doom_timer[doomed] -= accel[doomed]
    to_exec = doomed & (doom_timer <= 0)
    state[to_exec] = 2

    executing = (state == 2)
    exec_timer[executing] -= accel[executing]

def remove_completed(t_now):
    global pos, vel, clone_id, help_acc, div_clock, state, doom_timer, exec_timer
    global cell_uid, parent_uid, birth_t, death_t
    done = (state == 2) & (exec_timer <= 0)
    if not np.any(done):
        return

    # record death times for removed cells
    removed_uids = cell_uid[done]
    for u in removed_uids:
        G_death[int(u)] = int(t_now)

    keep = ~done
    pos = pos[keep]
    vel = vel[keep]
    clone_id = clone_id[keep]
    help_acc = help_acc[keep]
    div_clock = div_clock[keep]
    state = state[keep]
    doom_timer = doom_timer[keep]
    exec_timer = exec_timer[keep]

    cell_uid = cell_uid[keep]
    parent_uid = parent_uid[keep]
    birth_t = birth_t[keep]
    death_t = death_t[keep]

# -----------------------------
# Division with macro-jump
# -----------------------------
def division_step(P_i, P_mean, t_now):
    global pos, vel, clone_id, next_clone, help_acc, div_clock, state, doom_timer, exec_timer
    global cell_uid, parent_uid, birth_t, death_t
    global next_uid

    if len(pos) >= N_cap:
        return

    N = len(pos)
    can_div = (state != 2)

    speed = 1.0 + help_boost * (help_acc > help_threshold)
    div_clock[can_div] -= speed[can_div].astype(int)

    to_div = np.where((div_clock <= 0) & can_div)[0]
    if to_div.size == 0:
        return

    rng.shuffle(to_div)
    to_div = to_div[:max(1, int(0.08*N))]

    mean_help = float(np.mean(help_acc)) if N > 0 else 0.0

    for p in to_div:
        if len(pos) >= N_cap:
            break

        # daughter placement
        jitter = rng.normal(scale=r_cell*0.45, size=2)
        x_new = clip_to_balloon(pos[p] + jitter)
        v_new = vel[p] + rng.normal(scale=0.01, size=2)

        pos = np.vstack([pos, x_new[None,:]])
        vel = np.vstack([vel, v_new[None,:]])

        # macro-jump gating
        Pn = float(P_i[p] / (P_mean + 1e-12))
        Hn = float(help_acc[p] / (mean_help + 1e-12))

        jump_p = jump_base \
                 + jump_help_gain * sigmoid(jump_sigmoid_k * (1.0 - Hn)) \
                 + jump_press_gain * sigmoid(jump_sigmoid_k * (Pn - 1.0))
        jump_p = float(np.clip(jump_p, 0.0, 0.25))

        if rng.random() < jump_p and next_clone < max_clones:
            c_new = next_clone
            next_clone += 1
        else:
            c_new = int(clone_id[p])

        clone_id = np.append(clone_id, c_new)
        help_acc = np.append(help_acc, 0.0)

        div_clock[p] = int(rng.integers(div_min, div_max+1))
        div_clock = np.append(div_clock, int(rng.integers(div_min, div_max+1)))

        # apoptosis state new
        state = np.append(state, 0).astype(np.int8)
        doom_timer = np.append(doom_timer, 0.0)
        exec_timer = np.append(exec_timer, 0.0)

        # genealogy new
        uid_new = next_uid
        next_uid += 1

        cell_uid = np.append(cell_uid, uid_new)
        parent_uid = np.append(parent_uid, int(cell_uid[p]))  # parent is p's UID (after append ok because p index unchanged)
        birth_t = np.append(birth_t, int(t_now))
        death_t = np.append(death_t, -1)

        # global dicts
        G_parent[int(uid_new)] = int(cell_uid[p])
        G_birth[int(uid_new)] = int(t_now)
        G_clone[int(uid_new)] = int(c_new)

# -----------------------------
# Logging & snapshots
# -----------------------------
snap_idx = np.linspace(0, steps-1, snapshots, dtype=int).tolist()
snaps = []

P_series = []
N_series = []
dom_series = []
Neff_series = []
var_series = []

for t in range(steps):
    polys = clipped_voronoi(pos)
    gain = face_weighted_help(polys)
    help_acc = help_decay*help_acc + help_gain*gain

    pos, vel, P_i, P_mean = mechanics_step(pos, vel)

    # log
    P_series.append(P_mean)
    N_series.append(len(pos))
    dom_series.append(dom_fraction(clone_id))
    Neff_series.append(simpson_Neff(clone_id))
    if len(P_series) >= 25:
        var_series.append(float(np.var(P_series[-25:])))
    else:
        var_series.append(0.0)

    # alive -> doomed
    alive = np.where(state == 0)[0]
    for i in alive:
        p = death_hazard(P_i[i], P_mean, help_acc[i])
        if rng.random() < p:
            mark_doomed(i)

    # advance apoptosis & remove
    advance_apoptosis(P_i, P_mean)
    remove_completed(t)

    # division with macro-jumps
    division_step(P_i, P_mean, t)

    if t in snap_idx:
        snaps.append(dict(t=t, pos=pos.copy(), clone=clone_id.copy(), state=state.copy()))

# -----------------------------
# Save NPZ (timeseries + genealogy dicts)
# -----------------------------
npz_path = "/content/gc_balloon_advanced_with_jumps.npz"
np.savez_compressed(
    npz_path,
    P=np.array(P_series, float),
    N=np.array(N_series, int),
    dom=np.array(dom_series, float),
    Neff=np.array(Neff_series, float),
    varP=np.array(var_series, float),
    # final snapshot arrays
    final_pos=pos,
    final_clone=clone_id,
    final_state=state,
    # genealogy (store as object arrays)
    G_parent=np.array(list(G_parent.items()), dtype=object),
    G_birth=np.array(list(G_birth.items()), dtype=object),
    G_death=np.array(list(G_death.items()), dtype=object),
    G_clone=np.array(list(G_clone.items()), dtype=object),
    params=dict(
        R=R, r_cell=r_cell, steps=steps, dt=dt,
        div_min=div_min, div_max=div_max,
        doom_min=doom_min, doom_max=doom_max,
        exec_min=exec_min, exec_max=exec_max,
        jump_base=jump_base, jump_help_gain=jump_help_gain, jump_press_gain=jump_press_gain,
        pressure_accel=pressure_accel
    )
)
print("Saved NPZ:", npz_path)

# -----------------------------
# Plots: foam montage
# -----------------------------
rows, cols = 3, 4
fig, axes = plt.subplots(rows, cols, figsize=(10.8, 8.4), dpi=170, constrained_layout=True)
cmap = plt.get_cmap("tab20")

fx, fy = FDC.exterior.xy

for ax, fr in zip(axes.ravel(), snaps):
    ax.set_aspect("equal", "box")
    ax.set_xlim(-R, R)
    ax.set_ylim(-R, R)
    ax.axis("off")
    ax.add_patch(plt.Circle((0,0), R, fill=False, linewidth=0.8, edgecolor="black"))
    ax.plot(fx, fy, linewidth=0.6)

    Pp = fr["pos"]
    Cc = fr["clone"]
    Ss = fr["state"]
    polys = clipped_voronoi(Pp)

    for i, poly in enumerate(polys):
        if poly.is_empty:
            continue
        x, y = poly.exterior.xy
        col = cmap(int(Cc[i]) % 20)
        alpha = 0.55
        if Ss[i] == 1: alpha = 0.35
        if Ss[i] == 2: alpha = 0.18
        ax.fill(x, y, color=col, alpha=alpha, linewidth=0.25, edgecolor="black")

    ax.scatter(Pp[:,0], Pp[:,1], s=2, c="black", alpha=0.50)
    ax.set_title(f"t={fr['t']:03d} | N={len(Pp)} | dom={dom_fraction(Cc):.2f} | Neff={simpson_Neff(Cc):.1f}", fontsize=8)

for ax in axes.ravel()[len(snaps):]:
    ax.axis("off")

montage_path = "/content/gc_balloon_advanced_montage.png"
plt.savefig(montage_path, bbox_inches="tight")
plt.show()
print("Saved montage:", montage_path)

# -----------------------------
# Timeseries plots
# -----------------------------
t = np.arange(len(P_series))

def save_ts(y, title, ylabel, out):
    plt.figure(figsize=(10.4, 3.2), dpi=170)
    plt.plot(t, y, linewidth=1.2)
    plt.title(title)
    plt.xlabel("time step")
    plt.ylabel(ylabel)
    plt.tight_layout()
    plt.savefig(out, bbox_inches="tight")
    plt.show()
    print("Saved:", out)

save_ts(P_series, "Pressure proxy (mean overlap)", "pressure", "/content/gc_pressure.png")
save_ts(Neff_series, r"Effective manifold number $N_{\mathrm{eff}}$ (Simpson)", "Neff", "/content/gc_neff.png")
save_ts(dom_series, "Dominant clone fraction", "dominant fraction", "/content/gc_dominance.png")
save_ts(var_series, "Instability proxy: rolling variance of pressure", "Var(P) over last 25 steps", "/content/gc_varP.png")

# ============================================================
# Lineage "honeycomb branching" plot (dominant manifold)
# ============================================================

# Rebuild dicts from current G_* for safety
birth = dict(G_birth)
death = dict(G_death)  # some nodes missing -> alive at end
parent = dict(G_parent)
clone = dict(G_clone)

# pick dominant clone at final time among living cells
final_counts = {}
for c in clone_id:
    final_counts[int(c)] = final_counts.get(int(c), 0) + 1
dominant_clone = max(final_counts.items(), key=lambda kv: kv[1])[0]
print("Dominant manifold for lineage plot:", dominant_clone)

# collect all uids that ever belonged to that manifold
uids = [u for u,c in clone.items() if c == dominant_clone]
uid_set = set(uids)

# build children adjacency
children = {u: [] for u in uid_set}
roots = []
for u in uid_set:
    p = parent.get(u, -1)
    if p in uid_set:
        children[p].append(u)
    else:
        roots.append(u)

# Layout: time on y, x assigned by recursive 60° branching
# Use a simple dx that shrinks with depth; dy is time difference.
# We'll compute x via recursive rule: each node's children spread +/- dx(depth).
xpos = {}
depth = {}

def assign(u, x, d):
    xpos[u] = x
    depth[u] = d
    kids = children.get(u, [])
    if not kids:
        return
    # sort kids by birth time so layout stable
    kids = sorted(kids, key=lambda k: birth.get(k, 0))
    # branch step: use decreasing dx with depth
    base_dx = 0.60 * (0.72 ** d)
    # alternate left/right around parent
    offsets = []
    for i in range(len(kids)):
        sign = -1 if i % 2 == 0 else +1
        offsets.append(sign * base_dx * (1 + 0.12*i))
    for k, off in zip(kids, offsets):
        assign(k, x + off, d+1)

# Assign each root spaced apart
if len(roots) == 0:
    print("No roots found for dominant manifold (unexpected).")
else:
    roots = sorted(roots, key=lambda u: birth.get(u, 0))
    root_spacing = 2.0
    for i, r0 in enumerate(roots):
        assign(r0, i*root_spacing, 0)

# Build plot edges
plt.figure(figsize=(10.5, 5.4), dpi=170)
for u in uid_set:
    p = parent.get(u, -1)
    if p in uid_set:
        x0, y0 = xpos[p], birth.get(p, 0)
        x1, y1 = xpos[u], birth.get(u, 0)
        # draw as two 60° segments by using straight line; "honeycomb feel" comes from fixed offsets
        plt.plot([x0, x1], [y0, y1], linewidth=0.8, alpha=0.9)

# mark death times
for u in uid_set:
    if u in death:
        plt.scatter([xpos[u]], [death[u]], marker="x", s=18)

plt.gca().invert_yaxis()  # earlier time at top (optional)
plt.title(f"Lineage honeycomb plot (manifold {dominant_clone}) | x=branching, y=time | x marks = death")
plt.xlabel("branch coordinate (60°-style offsets)")
plt.ylabel("time step")
plt.grid(True, axis="x", linewidth=0.3, alpha=0.3)
lineage_path = "/content/gc_lineage_honeycomb.png"
plt.tight_layout()
plt.savefig(lineage_path, bbox_inches="tight")
plt.show()
print("Saved lineage plot:", lineage_path)

print("\nAll outputs are in /content:")
print(" -", npz_path)
print(" -", montage_path)
print(" - /content/gc_pressure.png")
print(" - /content/gc_neff.png")
print(" - /content/gc_dominance.png")
print(" - /content/gc_varP.png")
print(" -", lineage_path)
````

````python
import numpy as np
import matplotlib.pyplot as plt
import os, math
from matplotlib import animation

# ----------------------------
# Load NPZ
# ----------------------------
npz_path = "/content/gc_balloon_advanced_with_jumps.npz"  # adjust if needed
D = np.load(npz_path, allow_pickle=True)
print("Loaded:", npz_path)
print("Keys:", list(D.keys()))

def pairs_to_dict(arr):
    # arr is object array of (key,value) pairs
    out = {}
    for k, v in arr:
        out[int(k)] = int(v)
    return out

G_birth = pairs_to_dict(D["G_birth"])
G_clone = pairs_to_dict(D["G_clone"])
G_parent = pairs_to_dict(D["G_parent"])  # not needed for races, but useful later

# deaths dict may not include alive-at-end UIDs
G_death = {}
if "G_death" in D:
    for k, v in D["G_death"]:
        G_death[int(k)] = int(v)

T = int(len(D["P"]))  # simulation length (time steps)

# ----------------------------
# Reconstruct clone size trajectories n_k(t)
# from birth/death events (difference-array method)
# ----------------------------
uids = sorted(G_birth.keys())

# Determine clones present
clones = sorted(set(G_clone[u] for u in uids))
print("Unique manifolds:", len(clones), "range:", (min(clones), max(clones)))

# diff arrays per clone (sparse dict)
diff = {c: np.zeros(T + 1, dtype=int) for c in clones}

for u in uids:
    c = G_clone[u]
    tb = G_birth[u]
    td = G_death.get(u, T)  # if never died, alive until end
    tb = max(0, min(tb, T))
    td = max(0, min(td, T))
    if td <= tb:
        continue
    diff[c][tb] += 1
    diff[c][td] -= 1

# cumulative sum to get n_c(t)
counts = {c: np.cumsum(diff[c][:-1]) for c in clones}

# Compute dominant clone at each t
dom_clone = np.zeros(T, dtype=int)
dom_frac = np.zeros(T, dtype=float)
totalN = np.zeros(T, dtype=int)

for t in range(T):
    # total at t
    n_all = 0
    best_c = clones[0]
    best_n = -1
    for c in clones:
        n = int(counts[c][t])
        n_all += n
        if n > best_n:
            best_n = n
            best_c = c
    totalN[t] = n_all
    dom_clone[t] = best_c
    dom_frac[t] = (best_n / (n_all + 1e-12)) if n_all > 0 else 0.0

print("Total N(t) range:", int(totalN.min()), "to", int(totalN.max()))

# ----------------------------
# Moving-window “honeycomb races” movie
# X=time window, Y=raw clone size (autoscaled)
# Optional hex rendering via 60° transform:
#   x' = x + 0.5*y
#   y' = (sqrt(3)/2)*y
# ----------------------------
window = 140     # window length (time steps)
stride = 2       # how many time steps to slide per frame
max_frames = max(1, (T - window) // stride)

hex_style = True     # set False for ordinary time-vs-size plot
top_k = 10           # plot top K clones in the window (keeps it readable)
fade_others = False  # set True if you want light background traces

out_gif = "/content/gc_honeycomb_races.gif"   # change to mp4 if you prefer
fps = 16

# colormap for clone IDs
cmap = plt.get_cmap("tab20")

def hex_transform(x, y):
    # x, y arrays
    X = x + 0.5*y
    Y = (math.sqrt(3)/2.0)*y
    return X, Y

fig, ax = plt.subplots(figsize=(9.8, 4.6), dpi=160)
title = ax.set_title("")
ax.set_xlabel("time (window)")
ax.set_ylabel("clone size (autoscaled)")

lines = {}

def pick_top_clones(t0, t1):
    # average size within window
    means = []
    for c in clones:
        m = float(np.mean(counts[c][t0:t1]))
        means.append((m, c))
    means.sort(reverse=True)
    return [c for _, c in means[:top_k]]

def init():
    ax.clear()
    ax.set_xlabel("time (window)")
    ax.set_ylabel("clone size (autoscaled)")
    return []

def update(frame_idx):
    t0 = frame_idx * stride
    t1 = t0 + window
    if t1 > T:
        t1 = T
        t0 = max(0, T - window)

    ax.clear()

    # window time axis
    x = np.arange(t0, t1)
    xw = x - x[0]  # window-local time

    # choose which clones to show
    show = pick_top_clones(t0, t1)

    # autoscale y using shown clones
    ymax = 1
    for c in show:
        ymax = max(ymax, int(np.max(counts[c][t0:t1])))

    # plot background faint traces (optional)
    if fade_others:
        for c in clones:
            if c in show:
                continue
            y = counts[c][t0:t1].astype(float)
            if hex_style:
                X, Y = hex_transform(xw.astype(float), y)
                ax.plot(X, Y, linewidth=0.6, alpha=0.10, color="black")
            else:
                ax.plot(xw, y, linewidth=0.6, alpha=0.08, color="black")

    # plot selected clones
    for c in show:
        y = counts[c][t0:t1].astype(float)
        col = cmap(int(c) % 20)
        if hex_style:
            X, Y = hex_transform(xw.astype(float), y)
            ax.plot(X, Y, linewidth=2.0, color=col, alpha=0.95)
        else:
            ax.plot(xw, y, linewidth=2.0, color=col, alpha=0.95, label=f"{c}")

    # annotate dominant clone in this window
    c_dom = int(dom_clone[t1-1])
    f_dom = float(dom_frac[t1-1])
    N_now = int(totalN[t1-1])

    if hex_style:
        ax.set_xlabel("time (window)  [rendered on 60° lattice]")
        # y scale in transformed coords depends on ymax
        # set limits roughly based on ymax
        Xmax, Ymax = hex_transform(np.array([xw.max()], float), np.array([float(ymax)], float))
        ax.set_xlim(-5, float(Xmax[0]) + 5)
        ax.set_ylim(-5, float(Ymax[0]) + 5)
    else:
        ax.set_xlim(0, xw.max())
        ax.set_ylim(0, ymax * 1.15)
        ax.legend(loc="upper left", fontsize=8, frameon=False)

    ax.set_title(f"Honeycomb races | t=[{t0},{t1}) | N={N_now} | dom clone={c_dom} | dom frac={f_dom:.2f}")

    return []

ani = animation.FuncAnimation(fig, update, frames=max_frames, init_func=init, blit=False)

# Save as GIF (most reliable in Colab)
ani.save(out_gif, writer="pillow", fps=fps)
plt.close(fig)

print("Saved:", out_gif)
````

````python
import numpy as np
import matplotlib.pyplot as plt

# ====== Load your NPZ ======
npz_path = "/content/gc_balloon_advanced_with_jumps.npz"  # change if needed
D = np.load(npz_path, allow_pickle=True)
print("Loaded:", npz_path)
print("Keys:", list(D.keys()))
T = int(len(D["P"]))  # total time steps

def pairs_to_dict(arr):
    out = {}
    for k, v in arr:
        out[int(k)] = int(v)
    return out

G_birth = pairs_to_dict(D["G_birth"])
G_clone = pairs_to_dict(D["G_clone"])
G_parent = pairs_to_dict(D["G_parent"])

G_death = {}
if "G_death" in D:
    for k, v in D["G_death"]:
        G_death[int(k)] = int(v)

uids_all = sorted(G_birth.keys())

# ====== Choose a manifold to plot ======
# Default: dominant manifold by total "lifeline mass" (sum of lifetimes)
lifemass = {}
for u in uids_all:
    c = G_clone[u]
    tb = G_birth[u]
    td = G_death.get(u, T)
    lifemass[c] = lifemass.get(c, 0) + max(0, td - tb)

manifold = max(lifemass.items(), key=lambda kv: kv[1])[0]
print("Selected manifold:", manifold)

# Filter UIDs in that manifold
uids = [u for u in uids_all if G_clone[u] == manifold]
uid_set = set(uids)
print("Cells in manifold:", len(uids))

# Build children adjacency
children = {u: [] for u in uids}
roots = []
for u in uids:
    p = G_parent.get(u, -1)
    if p in uid_set:
        children[p].append(u)
    else:
        roots.append(u)

# Sort children by birth time for stable packing
for u in uids:
    children[u] = sorted(children[u], key=lambda x: G_birth.get(x, 0))
roots = sorted(roots, key=lambda x: G_birth.get(x, 0))

# ====== Packed y assignment (tree packing) ======
# We assign each node an interval in y; leaves get consecutive y slots.
# Parents sit at the mean of their leaves, siblings stay adjacent.
ypos = {}
leaf_order = []

def is_leaf(u):
    return len(children[u]) == 0

def dfs_assign(u):
    if is_leaf(u):
        leaf_order.append(u)
        ypos[u] = len(leaf_order) - 1
        return (ypos[u], ypos[u])
    lo = 10**9
    hi = -10**9
    for v in children[u]:
        a, b = dfs_assign(v)
        lo = min(lo, a)
        hi = max(hi, b)
    ypos[u] = 0.5 * (lo + hi)
    return (lo, hi)

# assign each root in sequence, separated by gaps
gap = 2.0
offset = 0.0
for r in roots:
    # temporarily assign, then shift by offset
    leaf_order_before = len(leaf_order)
    dfs_assign(r)
    # shift all new nodes in this root by offset
    new_nodes = leaf_order[leaf_order_before:]
    # Collect nodes in this root via stack
    stack = [r]
    root_nodes = []
    while stack:
        x = stack.pop()
        root_nodes.append(x)
        stack.extend(children[x])
    for u in root_nodes:
        ypos[u] = ypos[u] + offset
    offset = max(ypos[u] for u in root_nodes if u in ypos) + gap

# ====== Build lifelines ======
# Each cell is a horizontal segment from birth->death (or end of sim)
segments = []
for u in uids:
    tb = G_birth[u]
    td = G_death.get(u, T)
    if td <= tb:
        continue
    segments.append((tb, td, ypos[u], u))

# ====== Plot 1: barcode tree (individual cells) ======
# If too many cells, this can get heavy. Use density plot below instead.
max_draw = 5000  # safe cap for interactive plotting
dense_mode = len(segments) > max_draw

if not dense_mode:
    plt.figure(figsize=(12, 6), dpi=170)
    for tb, td, y, u in segments:
        plt.plot([tb, td], [y, y], linewidth=0.7)
    # mark death events
    for tb, td, y, u in segments:
        if u in G_death:
            plt.scatter([td], [y], marker="x", s=10)
    plt.title(f"Packed lineage barcode tree | manifold {manifold} | cells={len(segments)}")
    plt.xlabel("time step")
    plt.ylabel("packed branch coordinate (tree-compressed)")
    plt.tight_layout()
    out_png = "/content/gc_lineage_barcode.png"
    plt.savefig(out_png, bbox_inches="tight")
    plt.show()
    print("Saved:", out_png)

# ====== Plot 2: density barcode (for large lineages) ======
# Render as an image: y bins x time bins; intensity=alive count in that pixel.
# This is how you scale toward 10k–100k without drawing every segment.
time_bins = 520  # match your sim length (or lower for smoothing)
y_bins = 350     # choose based on how tall the packed tree is

# Determine y range
ymin = min(y for _, _, y, _ in segments) if segments else 0
ymax = max(y for _, _, y, _ in segments) if segments else 1
ymax = ymax + 1.0

# Rasterize
img = np.zeros((y_bins, time_bins), dtype=np.int32)

def y_to_bin(y):
    return int(np.clip((y - ymin) / (ymax - ymin + 1e-12) * (y_bins - 1), 0, y_bins - 1))

def t_to_bin(t):
    return int(np.clip(t / (T + 1e-12) * (time_bins - 1), 0, time_bins - 1))

for tb, td, y, u in segments:
    yb = y_to_bin(y)
    t0 = t_to_bin(tb)
    t1 = t_to_bin(td)
    if t1 <= t0:
        continue
    img[yb, t0:t1] += 1

plt.figure(figsize=(12, 6), dpi=170)
plt.imshow(img, origin="lower", aspect="auto")
plt.title(f"Lineage density barcode | manifold {manifold} | (scales to large N)")
plt.xlabel("time (binned)")
plt.ylabel("packed branch coordinate (binned)")
plt.tight_layout()
out_png2 = "/content/gc_lineage_density.png"
plt.savefig(out_png2, bbox_inches="tight")
plt.show()
print("Saved:", out_png2)
````

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import animation
import math

# deps
def ensure(pkg):
    try:
        __import__(pkg)
        return True
    except Exception:
        return False

if not ensure("scipy"):
    !pip -q install scipy

from scipy.ndimage import gaussian_filter

# ==========================
# Load NPZ
# ==========================
npz_path = "/content/gc_balloon_advanced_with_jumps.npz"  # adjust if needed
D = np.load(npz_path, allow_pickle=True)
T = int(len(D["P"]))
print("Loaded:", npz_path, "| T=", T)

def pairs_to_dict(arr):
    out = {}
    for k, v in arr:
        out[int(k)] = int(v)
    return out

G_birth = pairs_to_dict(D["G_birth"])
G_clone = pairs_to_dict(D["G_clone"])
G_parent = pairs_to_dict(D["G_parent"])
G_death = {}
for k, v in D["G_death"]:
    G_death[int(k)] = int(v)

uids_all = sorted(G_birth.keys())
clones = sorted(set(G_clone[u] for u in uids_all))

# ==========================
# Reconstruct clone counts over time: n_c(t)
# (event-driven diff arrays, then cumsum)
# ==========================
diff = {c: np.zeros(T + 1, dtype=np.int32) for c in clones}

for u in uids_all:
    c = G_clone[u]
    tb = G_birth[u]
    td = G_death.get(u, T)
    tb = max(0, min(tb, T))
    td = max(0, min(td, T))
    if td <= tb:
        continue
    diff[c][tb] += 1
    diff[c][td] -= 1

counts = {c: np.cumsum(diff[c][:-1]) for c in clones}

totalN = np.zeros(T, dtype=np.int32)
domN = np.zeros(T, dtype=np.int32)
domC = np.zeros(T, dtype=np.int32)

for t in range(T):
    best_c = clones[0]
    best_n = -1
    nsum = 0
    for c in clones:
        n = int(counts[c][t])
        nsum += n
        if n > best_n:
            best_n = n
            best_c = c
    totalN[t] = nsum
    domN[t] = best_n
    domC[t] = best_c

print("N range:", int(totalN.min()), "to", int(totalN.max()))
print("domN range:", int(domN.min()), "to", int(domN.max()))

# ==========================
# Packed-branch coordinate (Y): stable packing by clone blocks
# We assign each UID a y slot within its clone, stacked by clone.
# This is enough to create a “bolus” density movie (and scales).
# ==========================
uids_by_clone = {c: [] for c in clones}
for u in uids_all:
    uids_by_clone[G_clone[u]].append(u)

# stable order within clone: by birth time
for c in clones:
    uids_by_clone[c] = sorted(uids_by_clone[c], key=lambda u: (G_birth[u], u))

# assign y positions with gaps between clones
gap = 3
y_of_uid = {}
y_cursor = 0
for c in clones:
    for u in uids_by_clone[c]:
        y_of_uid[u] = y_cursor
        y_cursor += 1
    y_cursor += gap

Ymax = y_cursor + 1
print("Packed Ymax:", Ymax)

# ==========================
# Build full density image img[y, t]
# Each living UID contributes 1 across [birth, death)
# ==========================
img = np.zeros((Ymax, T), dtype=np.float32)
for u in uids_all:
    y = y_of_uid[u]
    tb = G_birth[u]
    td = G_death.get(u, T)
    tb = max(0, min(tb, T))
    td = max(0, min(td, T))
    if td <= tb:
        continue
    img[y, tb:td] += 1.0

# Smooth to get “bolus tissue” feel
# (more smoothing in Y than T gives the oesophagus/mouse-in-snake look)
img_s = gaussian_filter(img, sigma=(2.2, 1.2))

# ==========================
# Moving-window movie with Y autoscale (“zoom belly”)
# ==========================
window = 160     # time window width
stride = 3       # slide rate
fps = 16
out_gif = "/content/gc_bolus_honeycomb.gif"

# display scaling
pad_rows = 20   # extra rows above/below active region
eps = 1e-6

fig, ax = plt.subplots(figsize=(10.8, 5.0), dpi=160)

im = ax.imshow(img_s[:, :window], origin="lower", aspect="auto")
line_dom, = ax.plot([], [], linewidth=2.0)     # dominant clone trajectory
line_env, = ax.plot([], [], linewidth=2.0)     # belly envelope

ax.set_title("")
ax.set_xlabel("time (window)")
ax.set_ylabel("packed branch coordinate (auto-zoom)")

def init():
    im.set_data(img_s[:, :window])
    line_dom.set_data([], [])
    line_env.set_data([], [])
    return im, line_dom, line_env

def update(frame):
    t0 = frame * stride
    t1 = t0 + window
    if t1 > T:
        t1 = T
        t0 = max(0, T - window)

    sub = img_s[:, t0:t1]

    # auto-zoom rows: find active rows in this window
    row_activity = np.max(sub, axis=1)
    active = np.where(row_activity > eps)[0]
    if active.size == 0:
        y0, y1 = 0, min(Ymax, 200)
    else:
        y0 = max(int(active.min()) - pad_rows, 0)
        y1 = min(int(active.max()) + pad_rows, Ymax - 1)
        if y1 <= y0:
            y0, y1 = 0, min(Ymax, 200)

    sub2 = sub[y0:y1+1, :]

    im.set_data(sub2)
    im.set_extent([0, (t1 - t0), y0, y1])  # x in window coords, y in global packed coords

    # Rescale “belly envelope” and dom trajectory into [y0,y1]
    tt = np.arange(t0, t1)
    xw = tt - t0

    # scale by window max to fill vertical space (this is the “fat belly” effect)
    Nw = totalN[t0:t1].astype(float)
    Dw = domN[t0:t1].astype(float)

    Nmax = max(float(np.max(Nw)), 1.0)
    Dmax = max(float(np.max(Dw)), 1.0)

    # map to y range
    y_env = y0 + (Nw / Nmax) * (y1 - y0)
    y_dom = y0 + (Dw / Dmax) * (y1 - y0)

    # envelope (cyan) and dominant (white)
    line_env.set_data(xw, y_env)
    line_env.set_color("#00cfe6")  # cyan-ish
    line_env.set_alpha(0.9)

    line_dom.set_data(xw, y_dom)
    line_dom.set_color("white")
    line_dom.set_alpha(0.95)

    # cosmetics
    ax.set_xlim(0, (t1 - t0))
    ax.set_ylim(y0, y1)

    c_dom = int(domC[t1-1])
    frac_dom = float(domN[t1-1] / (totalN[t1-1] + 1e-12)) if totalN[t1-1] > 0 else 0.0
    ax.set_title(f"GC bolus | t=[{t0},{t1}) | N={int(totalN[t1-1])} | dom clone={c_dom} | dom frac={frac_dom:.2f}")

    return im, line_dom, line_env

frames = max(1, (T - window) // stride)
ani = animation.FuncAnimation(fig, update, frames=frames, init_func=init, blit=False)

ani.save(out_gif, writer="pillow", fps=fps)
plt.close(fig)

print("Saved:", out_gif)
````

````python
import math, os
import numpy as np
import matplotlib.pyplot as plt

# deps
def ensure(pkg):
    try:
        __import__(pkg)
        return True
    except Exception:
        return False

if not ensure("scipy"):
    !pip -q install scipy
if not ensure("shapely"):
    !pip -q install shapely

from scipy.spatial import Voronoi
from shapely.geometry import Polygon, Point
from matplotlib import animation

# =========================
# 0) Controls
# =========================
rng = np.random.default_rng(11)

# Macro plot / inset snapshot times (your dotted lines)
snap_times = [0, 80, 160, 240, 320, 400, 520]   # adjust as you like
T = 520                                         # total steps

# Movie controls
make_movie = True
movie_stride = 3          # take every 3rd step
fps = 16
movie_out = "/content/gc_envelope_with_insets.gif"  # gif is most reliable in Colab

# Scaling choice for macro curve:
# - "3D": radius ~ N^(1/3) (volume->radius)
# - "2D": radius ~ N^(1/2) (area->radius)
size_scaling = "3D"

# =========================
# 1) Foam sim parameters (same style as our advanced run)
# =========================
R0 = 1.0          # base GC balloon radius (used only for sim positions)
r_cell = 0.070
dt = 0.03
damp = 0.90

N_init = 60
N_cap = 260

k_rep = 70.0
k_wall = 150.0

div_min = 120
div_max = 180

doom_min = 160
doom_max = 240
exec_min = 20
exec_max = 40

pressure_accel = 1.2
death_base = 0.004
death_pressure_gain = 0.014

help_gain = 0.005
help_decay = 0.92
help_threshold = 0.35
help_boost = 0.55

FDC_center = np.array([0.55, 0.0])
FDC_radius = 0.35
FDC = Point(FDC_center[0], FDC_center[1]).buffer(FDC_radius)

n_start_clones = 10
max_clones = 120
new_clone_prob = 0.02  # keep small (you can increase later)

# =========================
# 2) Geometry helpers
# =========================
def circle_polygon(R, n=256):
    pts = [(R*math.cos(2*math.pi*i/n), R*math.sin(2*math.pi*i/n)) for i in range(n)]
    return Polygon(pts)

BALL0 = circle_polygon(R0, 256)

def rand_in_disc(R_inner):
    u = rng.random()
    ang = 2*np.pi*rng.random()
    rr = R_inner * np.sqrt(u)
    return np.array([rr*np.cos(ang), rr*np.sin(ang)], dtype=float)

def clip_to_balloon(x, R):
    n = np.linalg.norm(x)
    lim = R - r_cell
    if n <= lim:
        return x
    return x * (lim / (n + 1e-12))

def wall_force(x, R):
    n = np.linalg.norm(x) + 1e-12
    lim = R - r_cell
    if n <= lim:
        return np.zeros(2)
    pen = n - lim
    return -k_wall * pen * (x / n)

def voronoi_finite_polygons_2d(vor, radius=5.0):
    new_regions = []
    new_vertices = vor.vertices.tolist()
    center = vor.points.mean(axis=0)

    all_ridges = {}
    for (p1, p2), (v1, v2) in zip(vor.ridge_points, vor.ridge_vertices):
        all_ridges.setdefault(p1, []).append((p2, v1, v2))
        all_ridges.setdefault(p2, []).append((p1, v1, v2))

    for p1, reg_idx in enumerate(vor.point_region):
        verts = vor.regions[reg_idx]
        if all(v >= 0 for v in verts):
            new_regions.append(verts)
            continue

        ridges = all_ridges[p1]
        new_region = [v for v in verts if v >= 0]

        for p2, v1, v2 in ridges:
            if v1 < 0 or v2 < 0:
                v = v1 if v1 >= 0 else v2
                t = vor.points[p2] - vor.points[p1]
                t = t / (np.linalg.norm(t) + 1e-12)
                n = np.array([-t[1], t[0]])
                midpoint = (vor.points[p1] + vor.points[p2]) / 2
                direction = np.sign(np.dot(midpoint - center, n)) * n
                far_point = vor.vertices[v] + direction * radius
                new_vertices.append(far_point.tolist())
                new_region.append(len(new_vertices) - 1)

        vs = np.asarray([new_vertices[v] for v in new_region])
        c = vs.mean(axis=0)
        angles = np.arctan2(vs[:,1] - c[1], vs[:,0] - c[0])
        new_region = [v for _, v in sorted(zip(angles, new_region))]
        new_regions.append(new_region)

    return new_regions, np.asarray(new_vertices)

def clipped_voronoi(points, BALL):
    if len(points) < 4:
        polys = []
        for p in points:
            polys.append(Point(p[0], p[1]).buffer(r_cell*1.02).intersection(BALL))
        return polys

    vor = Voronoi(points)
    regions, vertices = voronoi_finite_polygons_2d(vor, radius=4.0)
    polys = []
    for reg in regions:
        poly = Polygon(vertices[reg]).intersection(BALL)
        if poly.is_empty or poly.area < 1e-6:
            poly = Point(vor.points[len(polys)][0], vor.points[len(polys)][1]).buffer(r_cell).intersection(BALL)
        polys.append(poly)
    return polys

def face_weighted_help(polys):
    gains = np.zeros(len(polys), dtype=float)
    for i, poly in enumerate(polys):
        inter = poly.intersection(FDC)
        if inter.is_empty:
            continue
        gains[i] = float(inter.length)
    return gains

# =========================
# 3) Run sim and store:
# - N(t)
# - snapshots at snap_times (pos, clone, state)
# =========================
pos = np.stack([rand_in_disc(R0*0.75) for _ in range(N_init)], axis=0)
vel = rng.normal(scale=0.01, size=pos.shape)

clone_id = rng.integers(0, n_start_clones, size=N_init)
next_clone = int(np.max(clone_id)) + 1

help_acc = np.zeros(N_init, dtype=float)
div_clock = rng.integers(div_min, div_max+1, size=N_init)

state = np.zeros(N_init, dtype=np.int8)   # 0 alive, 1 doomed, 2 executing
doom_timer = np.zeros(N_init, dtype=float)
exec_timer = np.zeros(N_init, dtype=float)

def mark_doomed(i):
    state[i] = 1
    doom_timer[i] = float(rng.integers(doom_min, doom_max+1))
    exec_timer[i] = float(rng.integers(exec_min, exec_max+1))

def death_hazard(P_i, P_mean, h):
    Pn = float(P_i / (P_mean + 1e-12))
    p = death_base + death_pressure_gain * max(Pn - 1.0, 0.0)
    p = p * (1.0 - 0.55*np.tanh(h))
    return float(np.clip(p, 0.0, 0.08))

def advance_apoptosis(P_i, P_mean):
    accel = 1.0 + pressure_accel * (P_i / (P_mean + 1e-12))
    accel = np.clip(accel, 0.5, 6.0)
    doomed = (state == 1)
    doom_timer[doomed] -= accel[doomed]
    to_exec = doomed & (doom_timer <= 0)
    state[to_exec] = 2
    executing = (state == 2)
    exec_timer[executing] -= accel[executing]

def remove_completed(BALL, t_now):
    global pos, vel, clone_id, help_acc, div_clock, state, doom_timer, exec_timer
    done = (state == 2) & (exec_timer <= 0)
    if not np.any(done):
        return
    keep = ~done
    pos = pos[keep]
    vel = vel[keep]
    clone_id = clone_id[keep]
    help_acc = help_acc[keep]
    div_clock = div_clock[keep]
    state = state[keep]
    doom_timer = doom_timer[keep]
    exec_timer = exec_timer[keep]

def mechanics_step(pos, vel, R_ball):
    N = len(pos)
    dx = pos[:,None,:] - pos[None,:,:]
    dist = np.linalg.norm(dx, axis=2) + 1e-12
    np.fill_diagonal(dist, 1e9)

    target = 2.0 * r_cell
    overlap = np.maximum(target - dist, 0.0)

    ux = dx[:,:,0] / dist
    uy = dx[:,:,1] / dist
    fx = k_rep * np.sum(overlap * ux, axis=1)
    fy = k_rep * np.sum(overlap * uy, axis=1)
    F = np.stack([fx, fy], axis=1)

    for i in range(N):
        F[i] += wall_force(pos[i], R_ball)

    vel = damp*vel + dt*F
    pos = pos + dt*vel
    for i in range(N):
        pos[i] = clip_to_balloon(pos[i], R_ball)

    P_i = np.sum(overlap, axis=1)
    P_mean = float(np.mean(P_i)) if N > 0 else 0.0
    return pos, vel, P_i, P_mean

def division_step(t_now):
    global pos, vel, clone_id, next_clone, help_acc, div_clock, state, doom_timer, exec_timer
    if len(pos) >= N_cap:
        return
    N = len(pos)
    can_div = (state != 2)
    speed = 1.0 + help_boost * (help_acc > help_threshold)
    div_clock[can_div] -= speed[can_div].astype(int)
    to_div = np.where((div_clock <= 0) & can_div)[0]
    if to_div.size == 0:
        return
    rng.shuffle(to_div)
    to_div = to_div[:max(1, int(0.08*N))]

    for p in to_div:
        if len(pos) >= N_cap:
            break
        jitter = rng.normal(scale=r_cell*0.45, size=2)
        x_new = clip_to_balloon(pos[p] + jitter, R0)  # positions live in base R0 coordinates
        v_new = vel[p] + rng.normal(scale=0.01, size=2)
        pos = np.vstack([pos, x_new[None,:]])
        vel = np.vstack([vel, v_new[None,:]])

        # occasional new clone label
        if rng.random() < new_clone_prob and next_clone < max_clones:
            c_new = next_clone
            next_clone += 1
        else:
            c_new = int(clone_id[p])

        clone_id = np.append(clone_id, c_new)
        help_acc = np.append(help_acc, 0.0)
        div_clock[p] = int(rng.integers(div_min, div_max+1))
        div_clock = np.append(div_clock, int(rng.integers(div_min, div_max+1)))
        state = np.append(state, 0).astype(np.int8)
        doom_timer = np.append(doom_timer, 0.0)
        exec_timer = np.append(exec_timer, 0.0)

# Run sim
N_series = np.zeros(T, dtype=int)
snapshots_store = {}

for t in range(T):
    polys = clipped_voronoi(pos, BALL0)
    gain = face_weighted_help(polys)
    help_acc = help_decay*help_acc + help_gain*gain

    pos, vel, P_i, P_mean = mechanics_step(pos, vel, R0)
    N_series[t] = len(pos)

    # alive->doomed
    alive = np.where(state == 0)[0]
    for i in alive:
        p = death_hazard(P_i[i], P_mean, help_acc[i])
        if rng.random() < p:
            mark_doomed(i)

    advance_apoptosis(P_i, P_mean)
    remove_completed(BALL0, t)
    division_step(t)

    if t in snap_times:
        snapshots_store[t] = dict(pos=pos.copy(), clone=clone_id.copy(), state=state.copy())

print("Sim complete. N range:", int(N_series.min()), "to", int(N_series.max()))

# =========================
# 4) Macro size curve: Diameter D(t)
# =========================
N0 = max(int(N_series[0]), 1)
if size_scaling.upper() == "3D":
    R_t = (N_series / N0) ** (1.0/3.0)
else:
    R_t = (N_series / N0) ** (1.0/2.0)

D_t = 2.0 * R_t  # diameter proxy (relative units)

# =========================
# 5) Static composite figure
# =========================
fig = plt.figure(figsize=(12.5, 5.5), dpi=170)
ax = fig.add_axes([0.06, 0.12, 0.64, 0.78])

ax.plot(np.arange(T), D_t, linewidth=2.0)
ax.set_xlabel("time step")
ax.set_ylabel("GC diameter proxy (relative)")
ax.set_title("GC size vs time with microscopic cross-sections (insets)")

# dotted lines
for tt in snap_times:
    ax.axvline(tt, linestyle="--", linewidth=0.8, alpha=0.45)

# inset helper
cmap = plt.get_cmap("tab20")
for i, tt in enumerate(snap_times):
    if tt not in snapshots_store:
        continue
    fr = snapshots_store[tt]
    P = fr["pos"]; C = fr["clone"]; S = fr["state"]
    BALL = BALL0  # drawing in base coordinates

    # place inset along top
    inset_w = 0.09
    inset_h = 0.30
    left = 0.72 + (i % 3) * (inset_w + 0.02)
    bottom = 0.62 - (i // 3) * (inset_h + 0.05)
    iax = fig.add_axes([left, bottom, inset_w, inset_h])
    iax.set_aspect("equal", "box")
    iax.axis("off")
    iax.set_xlim(-R0, R0); iax.set_ylim(-R0, R0)
    iax.add_patch(plt.Circle((0,0), R0, fill=False, linewidth=0.8))

    # draw foam
    polys = clipped_voronoi(P, BALL)
    for j, poly in enumerate(polys):
        if poly.is_empty:
            continue
        x, y = poly.exterior.xy
        col = cmap(int(C[j]) % 20)
        alpha = 0.55
        if S[j] == 1: alpha = 0.35
        if S[j] == 2: alpha = 0.18
        iax.fill(x, y, color=col, alpha=alpha, linewidth=0.2, edgecolor="black")
    iax.scatter(P[:,0], P[:,1], s=1.5, c="black", alpha=0.5)
    iax.set_title(f"t={tt}", fontsize=8)

out_static = "/content/gc_envelope_insets_static.png"
plt.savefig(out_static, bbox_inches="tight")
plt.show()
print("Saved static:", out_static)

# =========================
# 6) Movie (moving cursor + live inset)
# =========================
if make_movie:
    fig, ax = plt.subplots(figsize=(11.2, 4.6), dpi=160)
    ax.plot(np.arange(T), D_t, linewidth=2.0)
    ax.set_xlabel("time step")
    ax.set_ylabel("GC diameter proxy (relative)")
    ax.set_title("GC size vs time (cursor + live cross-section)")
    for tt in snap_times:
        ax.axvline(tt, linestyle="--", linewidth=0.8, alpha=0.35)

    cursor = ax.axvline(0, color="red", linewidth=1.5)

    # inset axis for live cross-section
    iax = fig.add_axes([0.73, 0.20, 0.24, 0.60])
    iax.set_aspect("equal", "box")
    iax.axis("off")
    iax.set_xlim(-R0, R0); iax.set_ylim(-R0, R0)

    def draw_cross_section(tnow):
        iax.clear()
        iax.set_aspect("equal", "box")
        iax.axis("off")
        iax.set_xlim(-R0, R0); iax.set_ylim(-R0, R0)
        iax.add_patch(plt.Circle((0,0), R0, fill=False, linewidth=0.9))
        # choose nearest stored snapshot
        tt = min(snapshots_store.keys(), key=lambda k: abs(k - tnow))
        fr = snapshots_store[tt]
        P = fr["pos"]; C = fr["clone"]; S = fr["state"]
        polys = clipped_voronoi(P, BALL0)
        for j, poly in enumerate(polys):
            if poly.is_empty:
                continue
            x, y = poly.exterior.xy
            col = cmap(int(C[j]) % 20)
            alpha = 0.55
            if S[j] == 1: alpha = 0.35
            if S[j] == 2: alpha = 0.18
            iax.fill(x, y, color=col, alpha=alpha, linewidth=0.2, edgecolor="black")
        iax.scatter(P[:,0], P[:,1], s=1.5, c="black", alpha=0.5)
        iax.set_title(f"microslice ~ t={tt}", fontsize=10)

    draw_cross_section(0)

    frames = list(range(0, T, movie_stride))

    def update(frame_t):
        cursor.set_xdata([frame_t, frame_t])
        draw_cross_section(frame_t)
        ax.set_title(f"GC size vs time | t={frame_t} | N={int(N_series[frame_t])}")
        return cursor,

    ani = animation.FuncAnimation(fig, update, frames=frames, blit=False, interval=1000//fps)

    # Save
    ani.save(movie_out, writer="pillow", fps=fps)
    plt.close(fig)
    print("Saved movie:", movie_out)
````

````python
import math
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import animation

# deps
def ensure(pkg):
    try:
        __import__(pkg)
        return True
    except Exception:
        return False

if not ensure("scipy"):
    !pip -q install scipy
if not ensure("shapely"):
    !pip -q install shapely

from scipy.spatial import Voronoi
from shapely.geometry import Polygon, Point

rng = np.random.default_rng(11)

# =========================
# Controls
# =========================
T = 520
fps = 18
stride = 1                       # 1 = smooth movie
movie_out = "/content/gc_size_with_true_bulging_inset.gif"

size_scaling = "3D"              # "3D" (N^(1/3)) or "2D" (N^(1/2))
R0 = 1.0
r_cell = 0.070

# =========================
# Use SAME sim params as before (copy from your working run)
# =========================
dt = 0.03
damp = 0.90
N_init = 60
N_cap = 260
k_rep = 70.0
k_wall = 150.0
div_min = 120
div_max = 180
doom_min = 160
doom_max = 240
exec_min = 20
exec_max = 40
pressure_accel = 1.2
death_base = 0.004
death_pressure_gain = 0.014
help_gain = 0.005
help_decay = 0.92
help_threshold = 0.35
help_boost = 0.55
FDC_center = np.array([0.55, 0.0])
FDC_radius = 0.35
FDC = Point(FDC_center[0], FDC_center[1]).buffer(FDC_radius)
n_start_clones = 10
max_clones = 120
new_clone_prob = 0.02

# =========================
# Geometry helpers
# =========================
def circle_polygon(R, n=256):
    pts = [(R*math.cos(2*math.pi*i/n), R*math.sin(2*math.pi*i/n)) for i in range(n)]
    return Polygon(pts)

BALL0 = circle_polygon(R0, 256)

def rand_in_disc(R_inner):
    u = rng.random()
    ang = 2*np.pi*rng.random()
    rr = R_inner * np.sqrt(u)
    return np.array([rr*np.cos(ang), rr*np.sin(ang)], dtype=float)

def clip_to_balloon(x, R):
    n = np.linalg.norm(x)
    lim = R - r_cell
    if n <= lim:
        return x
    return x * (lim / (n + 1e-12))

def wall_force(x, R):
    n = np.linalg.norm(x) + 1e-12
    lim = R - r_cell
    if n <= lim:
        return np.zeros(2)
    pen = n - lim
    return -k_wall * pen * (x / n)

def voronoi_finite_polygons_2d(vor, radius=5.0):
    new_regions = []
    new_vertices = vor.vertices.tolist()
    center = vor.points.mean(axis=0)
    all_ridges = {}
    for (p1, p2), (v1, v2) in zip(vor.ridge_points, vor.ridge_vertices):
        all_ridges.setdefault(p1, []).append((p2, v1, v2))
        all_ridges.setdefault(p2, []).append((p1, v1, v2))
    for p1, reg_idx in enumerate(vor.point_region):
        verts = vor.regions[reg_idx]
        if all(v >= 0 for v in verts):
            new_regions.append(verts); continue
        ridges = all_ridges[p1]
        new_region = [v for v in verts if v >= 0]
        for p2, v1, v2 in ridges:
            if v1 < 0 or v2 < 0:
                v = v1 if v1 >= 0 else v2
                t = vor.points[p2] - vor.points[p1]
                t = t / (np.linalg.norm(t) + 1e-12)
                n = np.array([-t[1], t[0]])
                midpoint = (vor.points[p1] + vor.points[p2]) / 2
                direction = np.sign(np.dot(midpoint - center, n)) * n
                far_point = vor.vertices[v] + direction * radius
                new_vertices.append(far_point.tolist())
                new_region.append(len(new_vertices) - 1)
        vs = np.asarray([new_vertices[v] for v in new_region])
        c = vs.mean(axis=0)
        angles = np.arctan2(vs[:,1] - c[1], vs[:,0] - c[0])
        new_region = [v for _, v in sorted(zip(angles, new_region))]
        new_regions.append(new_region)
    return new_regions, np.asarray(new_vertices)

def clipped_voronoi(points, BALL):
    if len(points) < 4:
        polys = []
        for p in points:
            polys.append(Point(p[0], p[1]).buffer(r_cell*1.02).intersection(BALL))
        return polys
    vor = Voronoi(points)
    regions, vertices = voronoi_finite_polygons_2d(vor, radius=4.0)
    polys = []
    for reg in regions:
        poly = Polygon(vertices[reg]).intersection(BALL)
        if poly.is_empty or poly.area < 1e-6:
            poly = Point(vor.points[len(polys)][0], vor.points[len(polys)][1]).buffer(r_cell).intersection(BALL)
        polys.append(poly)
    return polys

def face_weighted_help(polys):
    gains = np.zeros(len(polys), dtype=float)
    for i, poly in enumerate(polys):
        inter = poly.intersection(FDC)
        if not inter.is_empty:
            gains[i] = float(inter.length)
    return gains

# =========================
# Simulation (store every frame)
# =========================
pos = np.stack([rand_in_disc(R0*0.75) for _ in range(N_init)], axis=0)
vel = rng.normal(scale=0.01, size=pos.shape)
clone_id = rng.integers(0, n_start_clones, size=N_init)
next_clone = int(np.max(clone_id)) + 1

help_acc = np.zeros(N_init, dtype=float)
div_clock = rng.integers(div_min, div_max+1, size=N_init)
state = np.zeros(N_init, dtype=np.int8)
doom_timer = np.zeros(N_init, dtype=float)
exec_timer = np.zeros(N_init, dtype=float)

def mark_doomed(i):
    state[i] = 1
    doom_timer[i] = float(rng.integers(doom_min, doom_max+1))
    exec_timer[i] = float(rng.integers(exec_min, exec_max+1))

def death_hazard(P_i, P_mean, h):
    Pn = float(P_i / (P_mean + 1e-12))
    p = death_base + death_pressure_gain * max(Pn - 1.0, 0.0)
    p = p * (1.0 - 0.55*np.tanh(h))
    return float(np.clip(p, 0.0, 0.08))

def advance_apoptosis(P_i, P_mean):
    accel = 1.0 + pressure_accel * (P_i / (P_mean + 1e-12))
    accel = np.clip(accel, 0.5, 6.0)
    doomed = (state == 1)
    doom_timer[doomed] -= accel[doomed]
    to_exec = doomed & (doom_timer <= 0)
    state[to_exec] = 2
    executing = (state == 2)
    exec_timer[executing] -= accel[executing]

def remove_completed():
    global pos, vel, clone_id, help_acc, div_clock, state, doom_timer, exec_timer
    done = (state == 2) & (exec_timer <= 0)
    if not np.any(done):
        return
    keep = ~done
    pos = pos[keep]; vel = vel[keep]; clone_id = clone_id[keep]
    help_acc = help_acc[keep]; div_clock = div_clock[keep]
    state = state[keep]; doom_timer = doom_timer[keep]; exec_timer = exec_timer[keep]

def mechanics_step(pos, vel, R_ball):
    N = len(pos)
    dx = pos[:,None,:] - pos[None,:,:]
    dist = np.linalg.norm(dx, axis=2) + 1e-12
    np.fill_diagonal(dist, 1e9)
    target = 2.0 * r_cell
    overlap = np.maximum(target - dist, 0.0)
    ux = dx[:,:,0] / dist; uy = dx[:,:,1] / dist
    fx = k_rep * np.sum(overlap * ux, axis=1)
    fy = k_rep * np.sum(overlap * uy, axis=1)
    F = np.stack([fx, fy], axis=1)
    for i in range(N):
        F[i] += wall_force(pos[i], R_ball)
    vel = damp*vel + dt*F
    pos = pos + dt*vel
    for i in range(N):
        pos[i] = clip_to_balloon(pos[i], R_ball)
    P_i = np.sum(overlap, axis=1)
    P_mean = float(np.mean(P_i)) if N > 0 else 0.0
    return pos, vel, P_i, P_mean

def division_step():
    global pos, vel, clone_id, next_clone, help_acc, div_clock, state, doom_timer, exec_timer
    if len(pos) >= N_cap:
        return
    N = len(pos)
    can_div = (state != 2)
    speed = 1.0 + help_boost * (help_acc > help_threshold)
    div_clock[can_div] -= speed[can_div].astype(int)
    to_div = np.where((div_clock <= 0) & can_div)[0]
    if to_div.size == 0:
        return
    rng.shuffle(to_div)
    to_div = to_div[:max(1, int(0.08*N))]
    for p in to_div:
        if len(pos) >= N_cap:
            break
        jitter = rng.normal(scale=r_cell*0.45, size=2)
        x_new = clip_to_balloon(pos[p] + jitter, R0)
        v_new = vel[p] + rng.normal(scale=0.01, size=2)
        pos = np.vstack([pos, x_new[None,:]])
        vel = np.vstack([vel, v_new[None,:]])
        if rng.random() < new_clone_prob and next_clone < max_clones:
            c_new = next_clone
            next_clone += 1
        else:
            c_new = int(clone_id[p])
        clone_id = np.append(clone_id, c_new)
        help_acc = np.append(help_acc, 0.0)
        div_clock[p] = int(rng.integers(div_min, div_max+1))
        div_clock = np.append(div_clock, int(rng.integers(div_min, div_max+1)))
        state = np.append(state, 0).astype(np.int8)
        doom_timer = np.append(doom_timer, 0.0)
        exec_timer = np.append(exec_timer, 0.0)

# store full frames
frames = []
N_series = np.zeros(T, dtype=int)

for t in range(T):
    polys = clipped_voronoi(pos, BALL0)
    gain = face_weighted_help(polys)
    help_acc = help_decay*help_acc + help_gain*gain

    pos, vel, P_i, P_mean = mechanics_step(pos, vel, R0)
    N_series[t] = len(pos)

    alive = np.where(state == 0)[0]
    for i in alive:
        if rng.random() < death_hazard(P_i[i], P_mean, help_acc[i]):
            mark_doomed(i)

    advance_apoptosis(P_i, P_mean)
    remove_completed()
    division_step()

    # store only what we need for rendering
    frames.append((pos.copy(), clone_id.copy(), state.copy()))

print("Sim done. N range:", int(N_series.min()), "to", int(N_series.max()))

# =========================
# Macro diameter proxy D(t)
# =========================
N0 = max(int(N_series[0]), 1)
if size_scaling.upper() == "3D":
    R_t = (N_series / N0) ** (1.0/3.0)
else:
    R_t = (N_series / N0) ** (1.0/2.0)

D_t = 2.0 * R_t

# wide-angle camera radius = max bulge
R_cam = R0 * float(np.max(R_t)) * 1.05

# =========================
# Movie render: cursor + bulging inset
# =========================
cmap = plt.get_cmap("tab20")

fig, ax = plt.subplots(figsize=(11.2, 4.6), dpi=160)
ax.plot(np.arange(T), D_t, linewidth=2.0)
ax.set_xlabel("time step")
ax.set_ylabel("GC diameter proxy (relative)")
ax.set_title("GC size vs time (cursor + bulging microslice)")
cursor = ax.axvline(0, color="red", linewidth=1.5)

# inset axis: fixed wide-angle camera, but boundary & contents scale each frame
iax = fig.add_axes([0.73, 0.18, 0.24, 0.64])
iax.set_aspect("equal", "box")
iax.axis("off")
iax.set_xlim(-R_cam, R_cam)
iax.set_ylim(-R_cam, R_cam)

def draw_inset(tnow):
    iax.clear()
    iax.set_aspect("equal", "box")
    iax.axis("off")
    iax.set_xlim(-R_cam, R_cam)
    iax.set_ylim(-R_cam, R_cam)

    s = float(R_t[tnow])  # scale factor
    R_now = R0 * s

    # scaled balloon for drawing
    BALL_now = circle_polygon(R_now, 256)

    # scaled positions
    P, C, S = frames[tnow]
    P2 = P * s

    iax.add_patch(plt.Circle((0,0), R_now, fill=False, linewidth=1.0))

    polys = clipped_voronoi(P2, BALL_now)
    for j, poly in enumerate(polys):
        if poly.is_empty:
            continue
        x, y = poly.exterior.xy
        col = cmap(int(C[j]) % 20)
        alpha = 0.55
        if S[j] == 1: alpha = 0.35
        if S[j] == 2: alpha = 0.18
        iax.fill(x, y, color=col, alpha=alpha, linewidth=0.2, edgecolor="black")

    iax.scatter(P2[:,0], P2[:,1], s=1.5, c="black", alpha=0.5)
    iax.set_title(f"microslice t={tnow} | scale={s:.2f}", fontsize=10)

draw_inset(0)

frames_idx = list(range(0, T, stride))

def update(k):
    tnow = frames_idx[k]
    cursor.set_xdata([tnow, tnow])
    draw_inset(tnow)
    ax.set_title(f"GC size vs time | t={tnow} | N={int(N_series[tnow])} | D={D_t[tnow]:.2f}")
    return cursor,

ani = animation.FuncAnimation(fig, update, frames=len(frames_idx), blit=False, interval=1000//fps)
ani.save(movie_out, writer="pillow", fps=fps)
plt.close(fig)

print("Saved movie:", movie_out)
````

````python
import os, json, hashlib
import numpy as np
from datetime import datetime, UTC

RUN_BASENAME = "gc_multiI"
NPZ_PATH = f"{RUN_BASENAME}.npz"
JSON_PATH = f"{RUN_BASENAME}.json"
TMP_NPZ_PATH = f"{RUN_BASENAME}.tmp.npz"
TMP_JSON_PATH = f"{RUN_BASENAME}.tmp.json"

def sha256_file(path: str) -> str:
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            h.update(chunk)
    return h.hexdigest()

# ---------- YOU MUST PROVIDE THIS ----------
def simulate_one(*, seed: int, angle_deg: int, parameters: dict) -> np.ndarray:
    """
    Return N_series as a 1D int array of length parameters["steps"].
    Replace this with your actual simulation call.
    """
    # Example placeholder (delete): random walk-ish
    rng = np.random.default_rng(seed + angle_deg * 1000)
    steps = int(parameters["steps"])
    N = np.empty(steps, dtype=np.int64)
    N[0] = 60
    for t in range(1, steps):
        N[t] = max(0, N[t-1] + rng.integers(-1, 3))
    return N

def atomic_write_json(path: str, tmp_path: str, obj: dict):
    with open(tmp_path, "w") as f:
        json.dump(obj, f, indent=2)
    os.replace(tmp_path, path)

def atomic_write_npz(path: str, tmp_path: str, **arrays):
    np.savez(tmp_path, **arrays)
    os.replace(tmp_path, path)

def load_state():
    if os.path.exists(NPZ_PATH) and os.path.exists(JSON_PATH):
        npz = np.load(NPZ_PATH, allow_pickle=True)
        with open(JSON_PATH, "r") as f:
            meta = json.load(f)
        return npz, meta
    return None, None

def init_state(angles_deg, seeds, parameters):
    meta = {
        "created_utc": datetime.now(UTC).isoformat(),
        "angles_deg": list(map(int, angles_deg)),
        "seeds": list(map(int, seeds)),
        "parameters": parameters,
        "shape": "GC_Balloon_N_counts",
        "completed": []  # list of {"angle_deg":..., "seed":...}
    }
    atomic_write_json(JSON_PATH, TMP_JSON_PATH, meta)

    A = len(angles_deg)
    S = len(seeds)
    steps = int(parameters["steps"])

    counts = np.full((A, S, steps), fill_value=-1, dtype=np.int64)  # -1 = not yet run
    atomic_write_npz(NPZ_PATH, TMP_NPZ_PATH,
                     counts=counts,
                     angles=np.array(angles_deg, dtype=np.int64),
                     seeds=np.array(seeds, dtype=np.int64))
    return np.load(NPZ_PATH, allow_pickle=True), meta

def run_multiI(*, angles_deg, seeds, parameters, heartbeat_every=5):
    npz, meta = load_state()
    if npz is None:
        npz, meta = init_state(angles_deg, seeds, parameters)

    counts = npz["counts"]
    angles = npz["angles"]
    seeds_arr = npz["seeds"]

    completed_set = {(c["angle_deg"], c["seed"]) for c in meta.get("completed", [])}

    done = 0
    total = len(angles) * len(seeds_arr)

    for ai, ang in enumerate(angles.tolist()):
        for si, seed in enumerate(seeds_arr.tolist()):
            key = (ang, seed)
            if key in completed_set:
                done += 1
                continue

            N_series = simulate_one(seed=seed, angle_deg=int(ang), parameters=parameters)
            N_series = np.asarray(N_series, dtype=np.int64)
            if N_series.ndim != 1 or N_series.shape[0] != int(parameters["steps"]):
                raise ValueError(f"simulate_one returned shape {N_series.shape}, expected ({parameters['steps']},)")

            counts[ai, si, :] = N_series

            meta["completed"].append({"angle_deg": int(ang), "seed": int(seed),
                                      "N_min": int(N_series.min()), "N_max": int(N_series.max()),
                                      "saved_utc": datetime.now(UTC).isoformat()})

            # Atomic saves (resume-safe)
            atomic_write_npz(NPZ_PATH, TMP_NPZ_PATH,
                             counts=counts,
                             angles=angles,
                             seeds=seeds_arr)
            atomic_write_json(JSON_PATH, TMP_JSON_PATH, meta)

            completed_set.add(key)
            done += 1

            if done % heartbeat_every == 0 or done == total:
                last = meta["completed"][-1]
                print(f"[heartbeat] {done}/{total} done | angle={last['angle_deg']} seed={last['seed']} "
                      f"Nmin={last['N_min']} Nmax={last['N_max']}")

    # Final summary + hashes
    npz_hash = sha256_file(NPZ_PATH)
    json_hash = sha256_file(JSON_PATH)
    print("DONE.")
    print("NPZ:", NPZ_PATH, "SHA256:", npz_hash)
    print("JSON:", JSON_PATH, "SHA256:", json_hash)

# ---------- CONFIG ----------
parameters = {
    "R": 1.0,
    "r_cell": 0.07,
    "steps": 520,
    "dt": 0.03,
    "div_min": 120,
    "div_max": 180,
    "doom_min": 160,
    "doom_max": 240,
    "exec_min": 20,
    "exec_max": 40,
    "pressure_accel": 1.2
}

angles_deg = [0]            # set to [0, 30, 45, 60, 90] if you want multi-angle
seeds = list(range(10))     # multi-I: 10 initial conditions

run_multiI(angles_deg=angles_deg, seeds=seeds, parameters=parameters, heartbeat_every=2)
````

````python
import os, json, hashlib
import numpy as np
from datetime import datetime, UTC

SWEEP_NAME = "gc_param_sweep"
NPZ_PATH = f"{SWEEP_NAME}.npz"
JSON_PATH = f"{SWEEP_NAME}.json"
TMP_NPZ = f"{SWEEP_NAME}.tmp.npz"
TMP_JSON = f"{SWEEP_NAME}.tmp.json"

def atomic_write_json(path, tmp, obj):
    with open(tmp, "w") as f:
        json.dump(obj, f, indent=2)
    os.replace(tmp, path)

def atomic_write_npz(path, tmp, **arrays):
    np.savez(tmp, **arrays)
    os.replace(tmp, path)

def sample_params(rng):
    # Probabilities: sample 3 positive numbers then normalize
    w = rng.random(3)
    w = w / w.sum()
    p_help, p_fas, p_neglect = w.tolist()

    # Timescales in steps
    div_min = int(rng.integers(60, 181))
    div_max = int(rng.integers(max(div_min+1, 90), 261))
    doom_steps_to_die = int(rng.integers(40, 301))

    # Movement (step size / diffusion-ish)
    D_LZ = float(rng.uniform(0.3, 2.0))
    D_DZ = float(rng.uniform(0.3, 2.0))
    dz_drift = float(rng.uniform(0.0, 2.0))

    return {
        "p_help": p_help,
        "p_fas": p_fas,
        "p_neglect": p_neglect,
        "div_min": div_min,
        "div_max": div_max,
        "doom_steps_to_die": doom_steps_to_die,
        "D_LZ": D_LZ,
        "D_DZ": D_DZ,
        "dz_drift": dz_drift,
    }

# YOU provide this
def simulate_one(params, seed, base):
    """
    Return at least N_series: (T,) int.
    Optionally also return stats dict with extra metrics.
    """
    # Make params the single source of truth
    p_help = float(params["p_help"])
    p_fas = float(params["p_fas"])
    p_neg = float(params["p_neglect"])

    div_min = int(params["div_min"])
    div_max = int(params["div_max"])
    doom_steps_to_die = int(params["doom_steps_to_die"])

    D_LZ = float(params["D_LZ"])
    D_DZ = float(params["D_DZ"])
    dz_drift = float(params["dz_drift"])

    rng = np.random.default_rng(seed)
    T = int(base["steps"])
    N = np.empty(T, dtype=np.int64)

    current_N = int(base.get("N0", 60))
    N[0] = current_N

    for t in range(1, T):
        # Simple model for N change based on parameters
        # Births: depends on div_min/div_max and p_help
        birth_rate = (1.0 / div_max) * (1.0 + p_help * rng.random())
        # Deaths: depends on doom_steps_to_die and p_fas, p_neglect
        death_rate = (1.0 / doom_steps_to_die) * (p_fas * rng.random() + p_neg * rng.random())

        num_births = rng.poisson(birth_rate * current_N)
        num_deaths = rng.poisson(death_rate * current_N)

        # Placeholder for movement effects (D_LZ, D_DZ, dz_drift) - can be used to modulate birth/death rates further
        # For this simple N model, we'll just have them implicitly affect population change

        current_N = max(0, current_N + num_births - num_deaths)
        N[t] = current_N

    stats = {}  # fill later
    return N, stats

def summarize_run(N):
    N = np.asarray(N, dtype=np.int64)
    peak = int(N.max())
    t_peak = int(N.argmax())
    final = int(N[-1])
    ratio = float(final / peak) if peak > 0 else 0.0
    auc = int(N.sum())
    extinct = bool(peak == 0 or final == 0)
    return {"N_peak": peak, "t_peak": t_peak, "N_final": final, "final_over_peak": ratio, "AUC": auc, "extinct": extinct}

def load_or_init(n_param_sets, seeds, base):
    if os.path.exists(NPZ_PATH) and os.path.exists(JSON_PATH):
        npz = np.load(NPZ_PATH, allow_pickle=True)
        with open(JSON_PATH, "r") as f:
            meta = json.load(f)
        return npz, meta

    meta = {
        "created_utc": datetime.now(UTC).isoformat(),
        "n_param_sets": int(n_param_sets),
        "seeds": list(map(int, seeds)),
        "base": base,
        "completed": 0,
        "param_table": [],
        "run_summaries": []
    }
    atomic_write_json(JSON_PATH, TMP_JSON, meta)

    # store only summaries initially; you can store full N later if you want
    # shape: (P, S, 5) -> peak, t_peak, final, ratio*1e6, auc
    P = int(n_param_sets)
    S = len(seeds)
    summary = np.full((P, S, 5), -1, dtype=np.int64)
    atomic_write_npz(NPZ_PATH, TMP_NPZ, summary=summary)

    npz = np.load(NPZ_PATH, allow_pickle=True)
    return npz, meta

def run_sweep(n_param_sets=200, seeds=range(5), base=None, rng_seed=123):
    if base is None:
        base = {"steps": 520, "N0": 60}

    seeds = list(seeds)
    npz, meta = load_or_init(n_param_sets, seeds, base)

    summary_arr = npz["summary"]
    P, S, _ = summary_arr.shape

    rng = np.random.default_rng(rng_seed)

    start_p = meta["completed"]
    for p in range(start_p, P):
        params = sample_params(rng)
        meta["param_table"].append(params)

        for si, seed in enumerate(seeds):
            N, extra = simulate_one(params=params, seed=int(seed), base=base)
            s = summarize_run(N)

            summary_arr[p, si, 0] = s["N_peak"]
            summary_arr[p, si, 1] = s["t_peak"]
            summary_arr[p, si, 2] = s["N_final"]
            summary_arr[p, si, 3] = int(round(s["final_over_peak"] * 1_000_000))
            summary_arr[p, si, 4] = s["AUC"]

        meta["completed"] = p + 1

        atomic_write_npz(NPZ_PATH, TMP_NPZ, summary=summary_arr)
        atomic_write_json(JSON_PATH, TMP_JSON, meta)

        if (p + 1) % 10 == 0 or (p + 1) == P:
            print(f"[heartbeat] param_sets {p+1}/{P} done")

    print("DONE sweep:", NPZ_PATH, JSON_PATH)

# Example call
base = {"steps": 520, "N0": 60}
run_sweep(n_param_sets=200, seeds=range(10), base=base, rng_seed=42)
````

````python
import json
import numpy as np

NPZ_PATH = "gc_param_sweep.npz"
JSON_PATH = "gc_param_sweep.json"

npz = np.load(NPZ_PATH, allow_pickle=True)
summary = npz["summary"]  # (P, S, 5): peak, t_peak, final, ratio*1e6, auc

with open(JSON_PATH, "r") as f:
    meta = json.load(f)

params_table = meta["param_table"]
P, S, K = summary.shape
assert K == 5

peak   = summary[:, :, 0].astype(np.float64)
tpeak  = summary[:, :, 1].astype(np.float64)
final  = summary[:, :, 2].astype(np.float64)
ratio  = (summary[:, :, 3].astype(np.float64) / 1_000_000.0)
auc    = summary[:, :, 4].astype(np.float64)

# Basic derived indicators
# (extinction proxy: final==0 or peak==0)
extinct = ((final <= 0) | (peak <= 0)).astype(np.float64)

# Aggregate over seeds
m_peak   = peak.mean(axis=1)
m_tpeak  = tpeak.mean(axis=1)
m_final  = final.mean(axis=1)
m_ratio  = ratio.mean(axis=1)
m_auc    = auc.mean(axis=1)
p_ext    = extinct.mean(axis=1)  # fraction of seeds extinct

# ---- Define "GC-like" targets ----
# 1) Peak should NOT be at the end (avoid runaway): mean t_peak < 0.9*T
T = meta["base"]["steps"]
runaway = (m_tpeak >= 0.9 * T).astype(np.float64)

# 2) Some contraction: final/peak < 0.9 (tune as you like)
contraction_ok = (m_ratio < 0.9).astype(np.float64)

# 3) Avoid trivial collapse: extinction fraction < 0.3
survival_ok = (p_ext < 0.3).astype(np.float64)

# ---- Score (lower is better) ----
# Penalize runaway, extinction, and lack of contraction.
score = (
    4.0 * runaway +
    6.0 * p_ext +
    3.0 * np.maximum(0.0, m_ratio - 0.8) +   # reward stronger contraction below 0.8
    1.0 * np.maximum(0.0, 0.2 - m_ratio)     # but avoid "too strong" contraction if you want
)

# Mask to only keep sets that pass basic gates
mask = (contraction_ok > 0) & (survival_ok > 0) & (runaway == 0)

idx = np.where(mask)[0]
idx_sorted = idx[np.argsort(score[idx])]

print("Total param sets:", P)
print("Passing mask:", len(idx_sorted))

topN = min(20, len(idx_sorted))
print("\nTop candidates:")
for rank, p in enumerate(idx_sorted[:topN], start=1):
    prm = params_table[p]
    print(
        f"#{rank:2d}  pset={p:3d}  score={score[p]:.3f}  "
        f"peak={m_peak[p]:.1f}  tpeak={m_tpeak[p]:.1f}/{T}  "
        f"final={m_final[p]:.1f}  ratio={m_ratio[p]:.3f}  p_ext={p_ext[p]:.2f}  "
        f"p_help={prm['p_help']:.3f} p_fas={prm['p_fas']:.3f} p_neg={prm['p_neglect']:.3f}  "
        f"doom={prm['doom_steps_to_die']} div=[{prm['div_min']},{prm['div_max']}]  "
        f"D_LZ={prm['D_LZ']:.2f} D_DZ={prm['D_DZ']:.2f} drift={prm['dz_drift']:.2f}"
    )
````

I will adjust the `simulate_one` function to increase the impact of `density_factor_death` on the death rate, aiming for a more pronounced contraction and a mid-run peak.



````python
import os, json, hashlib
import numpy as np
from datetime import datetime, UTC

SWEEP_NAME = "gc_param_sweep"
NPZ_PATH = f"{SWEEP_NAME}.npz"
JSON_PATH = f"{SWEEP_NAME}.json"
TMP_NPZ = f"{SWEEP_NAME}.tmp.npz"
TMP_JSON = f"{SWEEP_NAME}.tmp.json"

def atomic_write_json(path, tmp, obj):
    with open(tmp, "w") as f:
        json.dump(obj, f, indent=2)
    os.replace(tmp, path)

def atomic_write_npz(path, tmp, **arrays):
    np.savez(tmp, **arrays)
    os.replace(tmp, path)

def sample_params(rng):
    # Probabilities: sample 3 positive numbers then normalize
    w = rng.random(3)
    w = w / w.sum()
    p_help, p_fas, p_neglect = w.tolist()

    # Timescales in steps
    div_min = int(rng.integers(60, 181))
    div_max = int(rng.integers(max(div_min+1, 90), 261))
    doom_steps_to_die = int(rng.integers(40, 301))

    # Movement (step size / diffusion-ish)
    D_LZ = float(rng.uniform(0.3, 2.0))
    D_DZ = float(rng.uniform(0.3, 2.0))
    dz_drift = float(rng.uniform(0.0, 2.0))

    # Encounter gating probability
    p_contact = float(rng.uniform(0.0, 1.0))

    # Soft carrying capacity
    K_capacity = int(rng.integers(100, 1000)) # Added K_capacity sampling

    return {
        "p_help": p_help,
        "p_fas": p_fas,
        "p_neglect": p_neglect,
        "div_min": div_min,
        "div_max": div_max,
        "doom_steps_to_die": doom_steps_to_die,
        "D_LZ": D_LZ,
        "D_DZ": D_DZ,
        "dz_drift": dz_drift,
        "p_contact": p_contact,
        "K_capacity": K_capacity, # Added K_capacity to return dict
    }

# YOU provide this
def simulate_one(params, seed, base):
    """
    Return at least N_series: (T,) int.
    Optionally also return stats dict with extra metrics.
    """
    # Make params the single source of truth
    p_help = float(params["p_help"])
    p_fas = float(params["p_fas"])
    p_neg = float(params["p_neglect"])

    div_min = int(params["div_min"])
    div_max = int(params["div_max"])
    doom_steps_to_die = int(params["doom_steps_to_die"])

    D_LZ = float(params["D_LZ"])
    D_DZ = float(params["D_DZ"])
    dz_drift = float(params["dz_drift"])
    p_contact = float(params["p_contact"])
    K_capacity = int(params["K_capacity"]) # Retrieve K_capacity

    rng = np.random.default_rng(seed)
    T = int(base["steps"])
    N = np.empty(T, dtype=np.int64)

    current_N = int(base.get("N0", 60))
    N[0] = current_N

    for t in range(1, T):
        # Density dependence factor
        # Approaches 0 as N approaches K, approaches 1 as N approaches 0
        density_factor_birth = max(0.0, 1.0 - current_N / (K_capacity + 1e-6))
        # Approaches 1 as N approaches K, approaches 0 as N approaches 0
        density_factor_death = min(1.0, current_N / (K_capacity + 1e-6))

        # Simple model for N change based on parameters
        # Births: depends on div_min/div_max, p_help, p_contact and K_capacity
        birth_rate = (1.0 / div_max) * (1.0 + p_help * rng.random() * p_contact) * density_factor_birth
        # Deaths: depends on doom_steps_to_die, p_fas, p_neglect, p_contact and K_capacity
        # Increased coefficient for density_factor_death to strengthen contraction
        death_rate = (1.0 / doom_steps_to_die) * (p_fas * rng.random() + p_neg * rng.random() * (1 - p_contact)) + 0.1 * density_factor_death # Increased density_factor_death coefficient

        num_births = rng.poisson(birth_rate * current_N)
        num_deaths = rng.poisson(death_rate * current_N)

        current_N = max(0, current_N + num_births - num_deaths)
        N[t] = current_N

    stats = {}  # fill later
    return N, stats

def summarize_run(N):
    N = np.asarray(N, dtype=np.int64)
    peak = int(N.max())
    t_peak = int(N.argmax())
    final = int(N[-1])
    ratio = float(final / peak) if peak > 0 else 0.0
    auc = int(N.sum())
    extinct = bool(peak == 0 or final == 0)
    return {"N_peak": peak, "t_peak": t_peak, "N_final": final, "final_over_peak": ratio, "AUC": auc, "extinct": extinct}

def load_or_init(n_param_sets, seeds, base):
    if os.path.exists(NPZ_PATH) and os.path.exists(JSON_PATH):
        npz = np.load(NPZ_PATH, allow_pickle=True)
        with open(JSON_PATH, "r") as f:
            meta = json.load(f)
        return npz, meta

    meta = {
        "created_utc": datetime.now(UTC).isoformat(),
        "n_param_sets": int(n_param_sets),
        "seeds": list(map(int, seeds)),
        "base": base,
        "completed": 0,
        "param_table": [],
        "run_summaries": []
    }
    atomic_write_json(JSON_PATH, TMP_JSON, meta)

    # store only summaries initially; you can store full N later if you want
    # shape: (P, S, 5) -> peak, t_peak, final, ratio*1e6, auc
    P = int(n_param_sets)
    S = len(seeds)
    summary = np.full((P, S, 5), -1, dtype=np.int64)
    atomic_write_npz(NPZ_PATH, TMP_NPZ, summary=summary)

    npz = np.load(NPZ_PATH, allow_pickle=True)
    return npz, meta

def run_sweep(n_param_sets=200, seeds=range(5), base=None, rng_seed=123):
    if base is None:
        base = {"steps": 520, "N0": 60}

    seeds = list(seeds)
    npz, meta = load_or_init(n_param_sets, seeds, base)

    summary_arr = npz["summary"]
    P, S, _ = summary_arr.shape

    rng = np.random.default_rng(rng_seed)

    start_p = meta["completed"]
    for p in range(start_p, P):
        params = sample_params(rng)
        meta["param_table"].append(params)

        for si, seed in enumerate(seeds):
            N, extra = simulate_one(params=params, seed=int(seed), base=base)
            s = summarize_run(N)

            summary_arr[p, si, 0] = s["N_peak"]
            summary_arr[p, si, 1] = s["t_peak"]
            summary_arr[p, si, 2] = s["N_final"]
            summary_arr[p, si, 3] = int(round(s["final_over_peak"] * 1_000_000))
            summary_arr[p, si, 4] = s["AUC"]

        meta["completed"] = p + 1

        atomic_write_npz(NPZ_PATH, TMP_NPZ, summary=summary_arr)
        atomic_write_json(JSON_PATH, TMP_JSON, meta)

        if (p + 1) % 10 == 0 or (p + 1) == P:
            print(f"[heartbeat] param_sets {p+1}/{P} done")

    print("DONE sweep:", NPZ_PATH, JSON_PATH)

# Example call
base = {"steps": 520, "N0": 60}
run_sweep(n_param_sets=200, seeds=range(10), base=base, rng_seed=42)
````

````python
import os, json, hashlib
import numpy as np
from datetime import datetime, UTC

SWEEP_NAME = "gc_param_sweep"
NPZ_PATH = f"{SWEEP_NAME}.npz"
JSON_PATH = f"{SWEEP_NAME}.json"
TMP_NPZ = f"{SWEEP_NAME}.tmp.npz"
TMP_JSON = f"{SWEEP_NAME}.tmp.json"

def atomic_write_json(path, tmp, obj):
    with open(tmp, "w") as f:
        json.dump(obj, f, indent=2)
    os.replace(tmp, path)

def atomic_write_npz(path, tmp, **arrays):
    np.savez(tmp, **arrays)
    os.replace(tmp, path)

def sample_params(rng):
    # Probabilities: sample 3 positive numbers then normalize
    w = rng.random(3)
    w = w / w.sum()
    p_help, p_fas, p_neglect = w.tolist()

    # Timescales in steps (1 hour = 15 steps)
    # div_min: 6 to 12 hours -> 90 to 180 steps
    div_min = int(rng.integers(90, 181))
    # div_max: 18 to 24 hours -> 270 to 360 steps (ensuring div_max > div_min)
    div_max = int(rng.integers(max(div_min+1, 270), 361))
    # doom_steps_to_die: 24 to 120 hours -> 360 to 1800 steps
    doom_steps_to_die = int(rng.integers(360, 1801))

    # Movement (step size / diffusion-ish) - retained existing ranges
    D_LZ = float(rng.uniform(0.3, 2.0))
    D_DZ = float(rng.uniform(0.3, 2.0))
    dz_drift = float(rng.uniform(0.0, 2.0))

    # Encounter gating probability - focused range 0.7-1.0
    p_contact = float(rng.uniform(0.7, 1.0))

    # Soft carrying capacity - focused range 200-600
    K_capacity = int(rng.integers(200, 601))

    return {
        "p_help": p_help,
        "p_fas": p_fas,
        "p_neglect": p_neglect,
        "div_min": div_min,
        "div_max": div_max,
        "doom_steps_to_die": doom_steps_to_die,
        "D_LZ": D_LZ,
        "D_DZ": D_DZ,
        "dz_drift": dz_drift,
        "p_contact": p_contact,
        "K_capacity": K_capacity,
    }

def simulate_one(params, seed, base):
    """
    Return the full N_series trajectory and placeholder values for max_LZ and max_DZ.
    Meaningful `max_LZ` and `max_DZ` values would require a more complex spatial model.
    """
    # Make params the single source of truth
    p_help = float(params["p_help"])
    p_fas = float(params["p_fas"])
    p_neg = float(params["p_neglect"])

    div_min = int(params["div_min"])
    div_max = int(params["div_max"])
    doom_steps_to_die = int(params["doom_steps_to_die"])

    D_LZ = float(params["D_LZ"])
    D_DZ = float(params["D_DZ"])
    dz_drift = float(params["dz_drift"])
    p_contact = float(params["p_contact"])
    K_capacity = int(params["K_capacity"]) # Retrieve K_capacity

    rng = np.random.default_rng(seed)
    T = int(base["steps"])
    N_series = np.empty(T, dtype=np.int64)

    current_N = int(base.get("N0", 60))
    N_series[0] = current_N

    for t in range(1, T):
        # Density dependence factor
        # Approaches 0 as N approaches K, approaches 1 as N approaches 0
        density_factor_birth = max(0.0, 1.0 - current_N / (K_capacity + 1e-6))
        # Approaches 1 as N approaches K, approaches 0 as N approaches 0
        density_factor_death = min(1.0, current_N / (K_capacity + 1e-6))

        # Simple model for N change based on parameters
        # Births: depends on div_min/div_max, p_help, p_contact and K_capacity
        birth_rate = (1.0 / div_max) * (1.0 + p_help * rng.random() * p_contact) * density_factor_birth
        # Deaths: depends on doom_steps_to_die, p_fas, p_neglect, p_contact and K_capacity
        # Increased coefficient for density_factor_death to strengthen contraction
        death_rate = (1.0 / doom_steps_to_die) * (p_fas * rng.random() + p_neg * rng.random() * (1 - p_contact)) + 0.1 * density_factor_death # Increased density_factor_death coefficient

        num_births = rng.poisson(birth_rate * current_N)
        num_deaths = rng.poisson(death_rate * current_N)

        current_N = max(0, current_N + num_births - num_deaths)
        N_series[t] = current_N

    # max_LZ and max_DZ would require a more complex spatial model to compute.
    # For this simplified population model, they are returned as placeholders (0.0).
    max_LZ = 0.0
    max_DZ = 0.0

    return N_series, max_LZ, max_DZ

def summarize_run(N_series, max_LZ=0.0, max_DZ=0.0):
    N = np.asarray(N_series, dtype=np.int64)
    peak = int(N.max())
    t_peak = int(N.argmax())
    final = int(N[-1])
    ratio = float(final / peak) if peak > 0 else 0.0
    auc = int(N.sum())
    extinct = bool(peak == 0 or final == 0)
    return {"N_peak": peak, "t_peak": t_peak, "N_final": final, "final_over_peak": ratio, "AUC": auc, "extinct": extinct, "max_LZ": max_LZ, "max_DZ": max_DZ}

def load_or_init(n_param_sets, seeds, base, force_reinit=False):
    if (not force_reinit) and os.path.exists(NPZ_PATH) and os.path.exists(JSON_PATH):
        npz = np.load(NPZ_PATH, allow_pickle=True)
        with open(JSON_PATH, "r") as f:
            meta = json.load(f)
        return npz, meta

    meta = {
        "created_utc": datetime.now(UTC).isoformat(),
        "n_param_sets": int(n_param_sets),
        "seeds": list(map(int, seeds)),
        "base": base,
        "completed": 0,
        "param_table": [],
        "run_summaries": []
    }
    atomic_write_json(JSON_PATH, TMP_JSON, meta)

    P = int(n_param_sets)
    S = len(seeds)
    T = int(base["steps"])

    # summary metrics (7):
    # 0 peak, 1 t_peak, 2 final, 3 ratio*1e6, 4 auc, 5 max_LZ*1e6, 6 max_DZ*1e6
    summary = np.full((P, S, 7), -1, dtype=np.int64)

    # store full trajectories
    all_N_series = np.full((P, S, T), -1, dtype=np.int64)

    atomic_write_npz(NPZ_PATH, TMP_NPZ, summary=summary, all_N_series=all_N_series)

    npz = np.load(NPZ_PATH, allow_pickle=True)
    return npz, meta


def run_sweep(n_param_sets=200, seeds=range(5), base=None, rng_seed=123, force_reinit=False):
    if base is None:
        base = {"steps": 520, "N0": 60}

    seeds = list(seeds)
    npz, meta = load_or_init(n_param_sets, seeds, base, force_reinit=force_reinit)

    summary_arr = npz["summary"]
    all_N_series = npz["all_N_series"]
    P, S, M = summary_arr.shape
    assert M == 7, f"Expected 7 metrics, got {M}"
    assert all_N_series.shape == (P, S, int(base["steps"]))

    rng = np.random.default_rng(rng_seed)

    start_p = meta["completed"]
    for p in range(start_p, P):
        params = sample_params(rng)
        meta["param_table"].append(params)

        for si, seed in enumerate(seeds):
            N_series, max_LZ, max_DZ = simulate_one(params=params, seed=int(seed), base=base)
            N_series = np.asarray(N_series, dtype=np.int64)
            if N_series.ndim != 1 or N_series.shape[0] != int(base["steps"]):
                raise ValueError(f"simulate_one returned shape {N_series.shape}, expected ({base['steps']},)")

            all_N_series[p, si, :] = N_series

            s = summarize_run(N_series, max_LZ=max_LZ, max_DZ=max_DZ)

            summary_arr[p, si, 0] = s["N_peak"]
            summary_arr[p, si, 1] = s["t_peak"]
            summary_arr[p, si, 2] = s["N_final"]
            summary_arr[p, si, 3] = int(round(s["final_over_peak"] * 1_000_000))
            summary_arr[p, si, 4] = s["AUC"]
            summary_arr[p, si, 5] = int(round(s["max_LZ"] * 1_000_000))
            summary_arr[p, si, 6] = int(round(s["max_DZ"] * 1_000_000))

        meta["completed"] = p + 1

        atomic_write_npz(NPZ_PATH, TMP_NPZ, summary=summary_arr, all_N_series=all_N_series)
        atomic_write_json(JSON_PATH, TMP_JSON, meta)

        if (p + 1) % 10 == 0 or (p + 1) == P:
            print(f"[heartbeat] param_sets {p+1}/{P} done")

    print("DONE sweep:", NPZ_PATH, JSON_PATH)
````
