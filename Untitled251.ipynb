```python
import numpy as np
import matplotlib.pyplot as plt

# --- CONSTANTS (eV, fs) ---
HBAR_EV_FS = 0.6582119569

# --- BASIC 3-STATE HAMILTONIAN BUILDER ---
def build_H(E_R, E_I, E_P, g_RI, g_IP):
    "Returns a 3x3 Hamiltonian in basis [R, I, P]. Energies, couplings in eV."
    H = np.zeros((3, 3), dtype=np.complex128)
    H[0, 0] = E_R
    H[1, 1] = E_I
    H[2, 2] = E_P
    H[0, 1] = H[1, 0] = g_RI
    H[1, 2] = H[2, 1] = g_IP
    return H

# --- EXAMPLE PARAMETERS ---
E_R  = 0.00
E_I  = 0.20
E_P  = -0.50
g_RI = 0.05
g_IP = 0.04

H = build_H(E_R, E_I, E_P, g_RI, g_IP)

# --- DIAGONALISE ONCE FOR FAST PROPAGATION ---
E_vals, V = np.linalg.eigh(H)
V_dag = np.conjugate(V.T)

def U_dt(dt_fs):
    phase = np.exp(-1j * E_vals * dt_fs / HBAR_EV_FS)
    return V @ np.diag(phase) @ V_dag

# --- TIME GRID ---
T_TOTAL_FS = 200.0
DT_FS      = 0.1
N_STEPS    = int(T_TOTAL_FS / DT_FS)

# --- INITIAL STATE: PURE REACTANT |R> = (1, 0, 0) ---
psi = np.zeros(3, dtype=np.complex128)
psi[0] = 1.0

# --- STORAGE ARRAYS ---
times_fs = np.zeros(N_STEPS + 1)
P_R = np.zeros(N_STEPS + 1)
P_I = np.zeros(N_STEPS + 1)
P_P = np.zeros(N_STEPS + 1)

# store t = 0
times_fs[0] = 0.0
P_R[0] = np.abs(psi[0])**2
P_I[0] = np.abs(psi[1])**2
P_P[0] = np.abs(psi[2])**2

# --- MAIN TDSE LOOP ---
U = U_dt(DT_FS)
for n in range(1, N_STEPS + 1):
    psi = U @ psi
    psi /= np.linalg.norm(psi)
    t = n * DT_FS
    times_fs[n] = t
    P_R[n] = np.abs(psi[0])**2
    P_I[n] = np.abs(psi[1])**2
    P_P[n] = np.abs(psi[2])**2

# --- PLOT ---
plt.figure(figsize=(6, 4))
plt.plot(times_fs, P_R, label='P(R)')
plt.plot(times_fs, P_I, label='P(I)')
plt.plot(times_fs, P_P, label='P(P)')
plt.xlabel('Time (fs)')
plt.ylabel('Population')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from numba import njit

HBAR_EV_FS = 0.6582119569  # ℏ in eV·fs

# ---------- JIT: single-site Lindblad evolution ----------

@njit
def evolve_site_jit(E_R, E_I, E_P,
                    g_RI, gamma_phi, Gamma_rel,
                    T_total_fs, dt_fs):
    """
    JIT-compiled evolution for a single site (3x3 rho).
    Returns final rho (3x3 complex).
    """

    # Hamiltonian H (3x3)
    H = np.zeros((3, 3), dtype=np.complex128)
    H[0, 0] = E_R
    H[1, 1] = E_I
    H[2, 2] = E_P
    H[0, 1] = g_RI
    H[1, 0] = g_RI

    # Lindblad operators
    L_phi = np.zeros((3, 3), dtype=np.complex128)
    L_phi[1, 1] = np.sqrt(gamma_phi)

    L_rel = np.zeros((3, 3), dtype=np.complex128)
    L_rel[2, 1] = np.sqrt(Gamma_rel)

    # initial rho = |R><R|
    rho = np.zeros((3, 3), dtype=np.complex128)
    rho[0, 0] = 1.0

    n_steps = int(T_total_fs / dt_fs)
    for _ in range(n_steps):
        # commutator [H, rho]
        Hrho = H @ rho
        rhoH = rho @ H
        comm = Hrho - rhoH
        drho = -1j / HBAR_EV_FS * comm

        # dephasing with L_phi
        L = L_phi
        L_dag = L.conj().T
        L_rho = L @ rho
        LrhoL = L_rho @ L_dag
        LdagL = L_dag @ L
        drho += LrhoL - 0.5 * (LdagL @ rho + rho @ LdagL)

        # relaxation with L_rel
        L = L_rel
        L_dag = L.conj().T
        L_rho = L @ rho
        LrhoL = L_rho @ L_dag
        LdagL = L_dag @ L
        drho += LrhoL - 0.5 * (LdagL @ rho + rho @ LdagL)

        rho = rho + dt_fs * drho

        # enforce Hermitian and trace 1
        rho = 0.5 * (rho + rho.conj().T)
        tr = rho[0, 0].real + rho[1, 1].real + rho[2, 2].real
        rho /= tr

    return rho

# ---------- Shield mapping κ → (g, γφ, Γrel) ----------

def params_from_shape_kappa(shape, kappa_eff):
    """
    Shape sets baseline gamma0, Gamma0.
    Shield: g^2 = κ * γφ with γφ ∝ κ.
    """
    if shape == "sphere":
        gamma0 = 0.002
        Gamma0 = 0.004
    elif shape == "cube":
        gamma0 = 0.0025
        Gamma0 = 0.0045
    elif shape == "rod":
        gamma0 = 0.003
        Gamma0 = 0.005
    elif shape == "plate":
        gamma0 = 0.002
        Gamma0 = 0.0042
    elif shape == "peanut":
        gamma0 = 0.003
        Gamma0 = 0.006
    elif shape == "stepped":
        gamma0 = 0.0035
        Gamma0 = 0.0065
    elif shape == "branched":
        gamma0 = 0.0032
        Gamma0 = 0.007
    else:
        gamma0 = 0.002
        Gamma0 = 0.004

    kappa_eff = np.clip(kappa_eff, 0.05, 2.5)

    gamma_phi = gamma0 * kappa_eff           # decoherence ∝ κ
    g_RI = kappa_eff * np.sqrt(gamma0)       # Shield: g^2 = κ γφ (S=1)
    Gamma_rel = Gamma0 * (0.5 + 0.5*kappa_eff)

    return g_RI, gamma_phi, Gamma_rel

def run_site(shape, kappa_eff, T_total_fs=150.0, dt_fs=0.3):
    """
    Python wrapper that calls JIT engine for one site.
    """
    E_R, E_I, E_P = 0.0, 0.10, -0.30
    g_RI, gamma_phi, Gamma_rel = params_from_shape_kappa(shape, kappa_eff)
    rho_final = evolve_site_jit(E_R, E_I, E_P,
                                g_RI, gamma_phi, Gamma_rel,
                                T_total_fs, dt_fs)
    return rho_final[2, 2].real  # product population

# ---------- Equal-mass random shapes, many particles ----------

shape_specs = {
    "sphere":   {"mu": 1.0, "sigma": 0.05},
    "cube":     {"mu": 0.8, "sigma": 0.10},
    "rod":      {"mu": 1.1, "sigma": 0.20},
    "plate":    {"mu": 0.7, "sigma": 0.10},
    "peanut":   {"mu": 1.2, "sigma": 0.25},
    "stepped":  {"mu": 1.3, "sigma": 0.30},
    "branched": {"mu": 1.4, "sigma": 0.35},
}

shapes = list(shape_specs.keys())

# you can tune these for more/less resolution and speed
N_SITES = 10   # surface sites per particle (same for all shapes → equal "mass")
N_REAL  = 10   # number of independent particles per shape
kappa_base_vals = np.linspace(0.2, 1.8, 15)

np.random.seed()

PP_mean = {s: [] for s in shapes}
PP_std  = {s: [] for s in shapes}

# warm up JIT once
_ = run_site("sphere", 1.0)

for k_base in kappa_base_vals:
    for shape in shapes:
        mu = shape_specs[shape]["mu"]
        sigma = shape_specs[shape]["sigma"]
        yields_particles = []

        for _ in range(N_REAL):
            # random curvature factors for this particle
            factors = np.random.normal(mu, sigma, N_SITES)
            factors = np.clip(factors, 0.1, 2.5)
            P_sites = []
            for f in factors:
                k_eff = k_base * f
                P_sites.append(run_site(shape, k_eff))
            yields_particles.append(np.mean(P_sites))

        yields_particles = np.array(yields_particles)
        PP_mean[shape].append(yields_particles.mean())
        PP_std[shape].append(yields_particles.std())

for s in shapes:
    PP_mean[s] = np.array(PP_mean[s])
    PP_std[s]  = np.array(PP_std[s])

# ---------- Plot with error bars ----------

colors = {
    "sphere":   "tab:blue",
    "cube":     "tab:orange",
    "rod":      "tab:green",
    "plate":    "tab:red",
    "peanut":   "tab:purple",
    "stepped":  "tab:brown",
    "branched": "tab:pink",
}

plt.figure(figsize=(8, 5))
for shape in shapes:
    plt.errorbar(
        kappa_base_vals,
        PP_mean[shape],
        yerr=PP_std[shape],
        label=shape,
        color=colors[shape],
        marker="o",
        markersize=3,
        linewidth=1,
        capsize=2,
    )

plt.xlabel("Base curvature κ_base (toy)")
plt.ylabel("Final product population P(P) (mean ± std)")
plt.title("Random equal-mass nanoparticles (JIT, 150 fs, dt=0.3)")
plt.legend(ncol=2)
plt.tight_layout()
plt.show()
```

```python
# ================================
#  Ni curvature + orbital figure
#  (NP κ + Ni κ + Ni HOMO cloud)
# ================================

!pip install -q pyscf

import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401
from pyscf import gto, scf
from pyscf.dft import numint

# -----------------------------
# 1. Nanoparticle curvature PDF (optional)
# -----------------------------

def sample_np_kappa(shape="peanut", k_base=1.2, N=20000):
    """
    Sample an effective curvature distribution κ for a model nanoparticle shape.
    Shapes defined via (mu, sigma) curvature factors.
    """
    shape_specs = {
        "sphere":   {"mu": 1.0, "sigma": 0.05},
        "cube":     {"mu": 0.8, "sigma": 0.10},
        "plate":    {"mu": 0.7, "sigma": 0.10},
        "rod":      {"mu": 1.1, "sigma": 0.20},
        "peanut":   {"mu": 1.2, "sigma": 0.25},
        "stepped":  {"mu": 1.3, "sigma": 0.30},
        "branched": {"mu": 1.4, "sigma": 0.35},
    }
    mu = shape_specs[shape]["mu"]
    sigma = shape_specs[shape]["sigma"]
    factors = np.random.normal(mu, sigma, N)
    factors = np.clip(factors, 0.1, 3.0)
    return k_base * factors

# -----------------------------
# 2. Build Ni atom
# -----------------------------

def build_ni_molecule():
    """
    Build a Ni atom with a reasonable all-electron basis.
    """
    mol = gto.Mole()
    mol.atom = "Ni  0.0  0.0  0.0"
    mol.basis = {"Ni": "def2-svp"}  # decent all-electron basis
    mol.charge = 0
    mol.spin = 0  # RHF closed-shell; not the true ground state, but fine for curvature shapes
    mol.build()
    return mol

# -----------------------------
# 3. Ni electron-curvature PDF
# -----------------------------

def ni_curvature_pdf(mol,
                     box=3.0,
                     ngrid=40,
                     rho_thresh=1e-4):
    """
    Compute Ni electron density and Laplacian, then extract |∇²ρ| as curvature PDF.
    """
    xs = np.linspace(-box, box, ngrid)
    ys = np.linspace(-box, box, ngrid)
    zs = np.linspace(-box, box, ngrid)
    X, Y, Z = np.meshgrid(xs, ys, zs, indexing="ij")
    coords = np.stack([X.ravel(), Y.ravel(), Z.ravel()], axis=1)

    mf = scf.RHF(mol)
    mf.kernel()

    ni = numint.NumInt()
    ao = ni.eval_ao(mol, coords)          # (N_grid, nao)
    dm = mf.make_rdm1()
    rho_flat = ni.eval_rho(mol, ao, dm)   # (N_grid,)
    rho = rho_flat.reshape((ngrid, ngrid, ngrid))

    dx = xs[1] - xs[0]
    lap = (
        np.gradient(np.gradient(rho, dx, axis=0), dx, axis=0) +
        np.gradient(np.gradient(rho, dx, axis=1), dx, axis=1) +
        np.gradient(np.gradient(rho, dx, axis=2), dx, axis=2)
    )

    mask = rho > rho_thresh
    kappa = np.abs(lap[mask].ravel())
    kappa = np.clip(kappa, 0, np.percentile(kappa, 99.5))
    return kappa, mf

# -----------------------------
# 4. Ni HOMO orbital cloud
# -----------------------------

def ni_homo_cloud(mol, mf,
                  box=3.0,
                  ngrid=40,
                  keep_percent=97.0,
                  max_points=4000):
    """
    Sample a 3D point cloud of the Ni HOMO orbital: points where |ψ|² is largest.
    Colour by sign of ψ (blue/red).
    """
    xs = np.linspace(-box, box, ngrid)
    ys = np.linspace(-box, box, ngrid)
    zs = np.linspace(-box, box, ngrid)
    X, Y, Z = np.meshgrid(xs, ys, zs, indexing="ij")
    coords = np.stack([X.ravel(), Y.ravel(), Z.ravel()], axis=1)

    ni = numint.NumInt()
    ao = ni.eval_ao(mol, coords)

    mo_coeff = mf.mo_coeff
    mo_occ   = mf.mo_occ
    occ_idx  = np.where(mo_occ > 1e-6)[0]
    homo_idx = occ_idx[-1]  # index of HOMO

    mo_vals = ao @ mo_coeff[:, homo_idx]
    mo_vals = np.real(mo_vals)
    prob = mo_vals**2

    thr = np.percentile(prob, keep_percent)
    mask = prob > thr
    coords_cloud = coords[mask]
    psi_cloud    = mo_vals[mask]

    if coords_cloud.shape[0] > max_points:
        idx = np.random.choice(coords_cloud.shape[0], max_points, replace=False)
        coords_cloud = coords_cloud[idx]
        psi_cloud    = psi_cloud[idx]

    return coords_cloud, psi_cloud

# -----------------------------
# 5. Combined figure generator
# -----------------------------

def make_ni_figure(include_np=True,
                   np_shape="peanut",
                   outfile="ni_curvature_figure.png"):
    """
    Make a 3-panel figure:
      (1) NP κ PDF (optional),
      (2) Ni curvature PDF (log-y),
      (3) Ni HOMO orbital cloud + Ni atom.
    """

    # NP curvature sample
    if include_np:
        kappa_np = sample_np_kappa(shape=np_shape)
    else:
        kappa_np = None

    # Ni atom calculations
    mol = build_ni_molecule()
    kappa_ni, mf = ni_curvature_pdf(mol)
    coords_cloud, psi_cloud = ni_homo_cloud(mol, mf)
    coords_atoms = mol.atom_coords()

    # layout
    if include_np:
        fig = plt.figure(figsize=(16, 4))
        ax1 = fig.add_subplot(1, 3, 1)
        ax2 = fig.add_subplot(1, 3, 2)
        ax3 = fig.add_subplot(1, 3, 3, projection='3d')
    else:
        fig = plt.figure(figsize=(12, 4))
        ax1 = fig.add_subplot(1, 2, 1)
        ax2 = fig.add_subplot(1, 2, 2, projection='3d')
        ax3 = None

    # Panel 1: NP κ PDF
    if include_np:
        ax1.hist(kappa_np, bins=60, density=True, alpha=0.9)
        ax1.set_xlabel("NP curvature κ (arb.)")
        ax1.set_ylabel("Probability density")
        ax1.set_title(f"NP κ PDF (shape: {np_shape})")

    # Panel 2: Ni curvature PDF (log-y)
    counts, edges = np.histogram(kappa_ni, bins=60, density=True)
    eps = 1e-12
    counts = np.maximum(counts, eps)
    centers = 0.5 * (edges[:-1] + edges[1:])

    if include_np:
        target_ax = ax2
    else:
        target_ax = ax1

    target_ax.bar(centers, counts,
                  width=(edges[1] - edges[0]),
                  color="orange", alpha=0.9)
    target_ax.set_yscale("log")
    target_ax.set_xlabel(r"Ni electron curvature $|\nabla^2 \rho|$ (arb.)")
    target_ax.set_ylabel("Probability density (log scale)")
    target_ax.set_title("Ni curvature PDF")

    # Panel 3: Ni HOMO cloud + atom
    if include_np:
        cloud_ax = ax3
    else:
        cloud_ax = ax2

    colors_cloud = np.where(psi_cloud >= 0, "blue", "red")
    cloud_ax.scatter(coords_cloud[:, 0],
                     coords_cloud[:, 1],
                     coords_cloud[:, 2],
                     s=4, c=colors_cloud, alpha=0.6)

    # Ni atom at origin
    cloud_ax.scatter([0], [0], [0],
                     s=120, color="grey", edgecolors="k")
    cloud_ax.text(0, 0, 0.3, "Ni", ha='center')

    cloud_ax.set_title("Ni HOMO orbital cloud")
    cloud_ax.set_xticks([]); cloud_ax.set_yticks([]); cloud_ax.set_zticks([])
    cloud_ax.set_box_aspect([1, 1, 1])
    lim = 2.5
    cloud_ax.set_xlim(-lim, lim)
    cloud_ax.set_ylim(-lim, lim)
    cloud_ax.set_zlim(-lim, lim)

    plt.tight_layout()
    plt.savefig(outfile, dpi=300)
    plt.show()
    print("Saved:", outfile)

# -----------------------------
# Run once to generate figure
# -----------------------------
make_ni_figure(include_np=True,
               np_shape="peanut",
               outfile="ni_curvature_figure.png")
```

```python
# =========================================
#  Fe–S cluster analysis from 6LK1.cif
#  - auto-detect Fe–S clusters
#  - build PySCF molecule for one cluster
#  - compute |∇²ρ| curvature PDF (orange comb)
#  - compute HOMO orbital cloud
#  - plot geometry + PDF + cloud (3 panels)
# =========================================

!pip install -q pyscf gemmi

import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401

import gemmi
from pyscf import gto, scf
from pyscf.dft import numint

# -----------------------------
# 1. Read CIF and find Fe/S atoms
# -----------------------------

cif_path = "6LK1.cif"  # ensure this is in your Colab working dir

doc = gemmi.cif.read_file(cif_path)
block = doc.sole_block()
structure = gemmi.make_structure_from_block(block)
model = structure[0]  # first model

# collect Fe and S atoms with indices
fes_atoms_all = []
for chain in model:
    for res in chain:
        for atom in res:
            el = atom.element.name
            if el in ("Fe", "FE", "S"):  # handle FE vs Fe
                fes_atoms_all.append({
                    "element": "Fe" if el.upper().startswith("FE") else "S",
                    "x": atom.pos.x,
                    "y": atom.pos.y,
                    "z": atom.pos.z,
                })

fes_coords = np.array([[a["x"], a["y"], a["z"]] for a in fes_atoms_all])
fes_elems  = np.array([a["element"] for a in fes_atoms_all])

print(f"Found {len(fes_atoms_all)} Fe/S atoms in 6LK1")

# -----------------------------
# 2. Cluster Fe/S atoms into Fe–S centers
#    via distance-based connectivity
# -----------------------------

def cluster_fes(coords, cutoff=3.0):
    """
    Simple graph clustering: connect atoms within cutoff; return connected components.
    coords: (N,3)
    """
    N = coords.shape[0]
    adj = [[] for _ in range(N)]
    for i in range(N):
        for j in range(i+1, N):
            d = np.linalg.norm(coords[i] - coords[j])
            if d < cutoff:
                adj[i].append(j)
                adj[j].append(i)
    visited = [False]*N
    clusters = []
    for i in range(N):
        if not visited[i]:
            stack = [i]
            comp = []
            visited[i] = True
            while stack:
                k = stack.pop()
                comp.append(k)
                for nb in adj[k]:
                    if not visited[nb]:
                        visited[nb] = True
                        stack.append(nb)
            clusters.append(comp)
    return clusters

clusters = cluster_fes(fes_coords, cutoff=3.0)
print(f"Found {len(clusters)} Fe/S-connected clusters")

for idx, comp in enumerate(clusters):
    n_fe = np.sum(fes_elems[comp] == "Fe")
    n_s  = np.sum(fes_elems[comp] == "S")
    print(f"  Cluster {idx}: {len(comp)} atoms ({n_fe} Fe, {n_s} S)")





# -----------------------------
# 3. Choose which cluster to analyse
# -----------------------------
# Pick a cluster index manually; often the first or second will be a nice Fe–S center.

cluster_idx = 0  # <-- change this if needed after inspecting printout

atom_indices = clusters[cluster_idx]
print("\nUsing cluster", cluster_idx, "with atoms:")
for i in atom_indices:
    print(f"  {i}: {fes_elems[i]}, ({fes_coords[i,0]:.3f}, {fes_coords[i,1]:.3f}, {fes_coords[i,2]:.3f})")





# -----------------------------
# 3. Choose which cluster to analyse
# -----------------------------
cluster_idx = 1  # <-- pick 1 or 5, not 0

atom_indices = clusters[cluster_idx]
print("\nUsing cluster", cluster_idx, "with atoms:")
for i in atom_indices:
    print(f"  {i}: {fes_elems[i]}, ({fes_coords[i,0]:.3f}, {fes_coords[i,1]:.3f}, {fes_coords[i,2]:.3f})")

# -----------------------------
# 4. Build PySCF molecule, centred at origin
# -----------------------------

cluster_coords = fes_coords[atom_indices]         # (n_atoms, 3)
center = cluster_coords.mean(axis=0)              # centre of mass / mean position

atom_lines = []
for i in atom_indices:
    sym = fes_elems[i]
    x, y, z = fes_coords[i] - center              # shift so cluster is around (0,0,0)
    atom_lines.append(f"{sym} {x:.6f} {y:.6f} {z:.6f}")

atom_str = "\n".join(atom_lines)
print("\nPySCF atom block (centred):\n", atom_str)

mol = gto.Mole()
mol.atom = atom_str
mol.basis = 'sto-3g'
mol.charge = 0
mol.spin = 0
mol.build()
# -----------------------------
# 5. Compute electron curvature PDF: |∇²ρ|
# -----------------------------

def cluster_curvature_pdf(mol,
                          box=4.0,
                          ngrid=40,
                          rho_thresh=1e-3):
    xs = np.linspace(-box, box, ngrid)
    ys = np.linspace(-box, box, ngrid)
    zs = np.linspace(-box, box, ngrid)
    X, Y, Z = np.meshgrid(xs, ys, zs, indexing="ij")
    coords = np.stack([X.ravel(), Y.ravel(), Z.ravel()], axis=1)

    mf = scf.RHF(mol)
    mf.kernel()

    ni = numint.NumInt()
    ao = ni.eval_ao(mol, coords)
    dm = mf.make_rdm1()
    rho_flat = ni.eval_rho(mol, ao, dm)
    rho = rho_flat.reshape((ngrid, ngrid, ngrid))

    dx = xs[1] - xs[0]
    lap = (
        np.gradient(np.gradient(rho, dx, axis=0), dx, axis=0) +
        np.gradient(np.gradient(rho, dx, axis=1), dx, axis=1) +
        np.gradient(np.gradient(rho, dx, axis=2), dx, axis=2)
    )

    mask = rho > rho_thresh
    kappa = np.abs(lap[mask].ravel())
    kappa = np.clip(kappa, 0, np.percentile(kappa, 99.5))
    return kappa, mf

kappa_fes, mf = cluster_curvature_pdf(mol)

# -----------------------------
# 6. HOMO orbital cloud from 6LK1 Fe–S cluster
# -----------------------------

def cluster_homo_cloud(mol, mf,
                       box=4.0,
                       ngrid=40,
                       keep_percent=97.0,
                       max_points=5000):
    xs = np.linspace(-box, box, ngrid)
    ys = np.linspace(-box, box, ngrid)
    zs = np.linspace(-box, box, ngrid)
    X, Y, Z = np.meshgrid(xs, ys, zs, indexing="ij")
    coords = np.stack([X.ravel(), Y.ravel(), Z.ravel()], axis=1)

    ni = numint.NumInt()
    ao = ni.eval_ao(mol, coords)

    mo_coeff = mf.mo_coeff
    mo_occ   = mf.mo_occ
    occ_idx  = np.where(mo_occ > 1e-6)[0]
    homo_idx = occ_idx[-1]

    mo_vals = ao @ mo_coeff[:, homo_idx]
    mo_vals = np.real(mo_vals)
    prob = mo_vals**2

    thr = np.percentile(prob, keep_percent)
    mask = prob > thr
    coords_cloud = coords[mask]
    psi_cloud    = mo_vals[mask]

    if coords_cloud.shape[0] > max_points:
        idx = np.random.choice(coords_cloud.shape[0], max_points, replace=False)
        coords_cloud = coords_cloud[idx]
        psi_cloud    = psi_cloud[idx]

    return coords_cloud, psi_cloud

coords_cloud, psi_cloud = cluster_homo_cloud(mol, mf)

coords_atoms = mol.atom_coords()
symbols      = [mol.atom_symbol(i) for i in range(mol.natm)]

# -----------------------------
# 7. Plot 3-panel figure:
#    (a) Fe–S geometry
#    (b) curvature PDF (log-y)
#    (c) HOMO cloud
# -----------------------------

fig = plt.figure(figsize=(15, 4))

# Panel (a) Fe–S geometry
ax1 = fig.add_subplot(1, 3, 1, projection='3d')
color_atoms = {"Fe": "red", "S": "yellow"}
for (x, y, z), sym in zip(coords_atoms, symbols):
    ax1.scatter(x, y, z, s=80,
                color=color_atoms.get(sym, "grey"),
                edgecolors="k")
    ax1.text(x, y, z + 0.3, sym, ha='center', fontsize=9)

# draw bonds by distance
for i in range(len(coords_atoms)):
    for j in range(i+1, len(coords_atoms)):
        d = np.linalg.norm(coords_atoms[i] - coords_atoms[j])
        if d < 3.0:  # Å, rough Fe–S / Fe–Fe if close
            ax1.plot([coords_atoms[i,0], coords_atoms[j,0]],
                     [coords_atoms[i,1], coords_atoms[j,1]],
                     [coords_atoms[i,2], coords_atoms[j,2]],
                     color="k", linewidth=0.7)

ax1.set_title(f"Fe–S cluster from 6LK1 (cluster {cluster_idx})")
ax1.set_xticks([]); ax1.set_yticks([]); ax1.set_zticks([])
max_range = (coords_atoms.max(axis=0) - coords_atoms.min(axis=0)).max()
mid = coords_atoms.mean(axis=0)
for setter, m in zip([ax1.set_xlim, ax1.set_ylim, ax1.set_zlim], mid):
    setter(m - max_range/2, m + max_range/2)

# Panel (b) curvature PDF (log-y)
ax2 = fig.add_subplot(1, 3, 2)
counts, edges = np.histogram(kappa_fes, bins=60, density=True)
eps = 1e-12
counts = np.maximum(counts, eps)
centers = 0.5 * (edges[:-1] + edges[1:])

ax2.bar(centers, counts,
        width=(edges[1]-edges[0]),
        color="orange", alpha=0.9)
ax2.set_yscale("log")
ax2.set_xlabel(r"Electron curvature $|\nabla^2 \rho|$ (arb.)")
ax2.set_ylabel("Probability density (log scale)")
ax2.set_title("Fe–S curvature PDF\n(6LK1-derived cluster)")

# Panel (c) HOMO cloud
ax3 = fig.add_subplot(1, 3, 3, projection='3d')
colors_cloud = np.where(psi_cloud >= 0, "blue", "red")
ax3.scatter(coords_cloud[:,0],
            coords_cloud[:,1],
            coords_cloud[:,2],
            s=3, c=colors_cloud, alpha=0.6)
for (x, y, z), sym in zip(coords_atoms, symbols):
    ax3.scatter(x, y, z, s=60,
                color=color_atoms.get(sym, "grey"),
                edgecolors="k")
ax3.set_title("Fe–S HOMO orbital cloud (6LK1)")
ax3.set_xticks([]); ax3.set_yticks([]); ax3.set_zticks([])
for setter, m in zip([ax3.set_xlim, ax3.set_ylim, ax3.set_zlim], mid):
    setter(m - max_range/2, m + max_range/2)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # noqa
import gemmi
from pyscf import gto, dft
from pyscf.dft import numint

# =========================================
# 1. Load CIF and collect Fe/S atoms
# =========================================

cif_path = "6LK1.cif"  # adjust if needed

doc = gemmi.cif.read_file(cif_path)
block = doc.sole_block()
structure = gemmi.make_structure_from_block(block)
model = structure[0]  # first model

fes_atoms_all = []
for chain in model:
    for res in chain:
        for atom in res:
            el = atom.element.name
            if el.upper().startswith("FE") or el.upper() == "S":
                fes_atoms_all.append({
                    "element": "Fe" if el.upper().startswith("FE") else "S",
                    "x": atom.pos.x,
                    "y": atom.pos.y,
                    "z": atom.pos.z,
                })

fes_coords = np.array([[a["x"], a["y"], a["z"]] for a in fes_atoms_all])
fes_elems  = np.array([a["element"] for a in fes_atoms_all])

print(f"Found {len(fes_atoms_all)} Fe/S atoms in 6LK1")

# =========================================
# 2. Cluster Fe/S atoms by distance
# =========================================

def cluster_fes(coords, cutoff=3.0):
    N = coords.shape[0]
    adj = [[] for _ in range(N)]
    for i in range(N):
        for j in range(i+1, N):
            d = np.linalg.norm(coords[i] - coords[j])
            if d < cutoff:
                adj[i].append(j)
                adj[j].append(i)
    visited = [False]*N
    clusters = []
    for i in range(N):
        if not visited[i]:
            stack = [i]
            comp = []
            visited[i] = True
            while stack:
                k = stack.pop()
                comp.append(k)
                for nb in adj[k]:
                    if not visited[nb]:
                        visited[nb] = True
                        stack.append(nb)
            clusters.append(comp)
    return clusters

clusters = cluster_fes(fes_coords, cutoff=3.0)
print(f"Found {len(clusters)} Fe/S-connected clusters")
for idx, comp in enumerate(clusters):
    n_fe = np.sum(fes_elems[comp] == "Fe")
    n_s  = np.sum(fes_elems[comp] == "S")
    print(f"  Cluster {idx}: {len(comp)} atoms ({n_fe} Fe, {n_s} S)")

# =========================================
# 3. Build centred PySCF Mole for a cluster
# =========================================

def build_cluster_mol(cluster_idx):
    atom_indices = clusters[cluster_idx]
    print("\nUsing cluster", cluster_idx, "with atoms:")
    cluster_coords = fes_coords[atom_indices]
    center = cluster_coords.mean(axis=0)

    atom_lines = []
    for i in atom_indices:
        sym = fes_elems[i]
        x, y, z = fes_coords[i] - center
        atom_lines.append(f"{sym} {x:.6f} {y:.6f} {z:.6f}")
        print(f"  {i}: {sym}, ({fes_coords[i,0]:.3f}, {fes_coords[i,1]:.3f}, {fes_coords[i,2]:.3f})")

    atom_str = "\n".join(atom_lines)
    print("\nPySCF atom block (centred):\n", atom_str)

    mol = gto.Mole()
    mol.atom = atom_str
    mol.basis = 'sto-3g'   # can upgrade to def2-svp later
    mol.charge = 0
    mol.spin = 0           # RKS will handle spin-density if needed
    mol.build()
    return mol, np.array([line.split()[0] for line in atom_lines]), cluster_coords - center

# =========================================
# 4. Curvature PDF and HOMO cloud (using RKS)
# =========================================

def cluster_curvature_pdf(mol,
                          box=4.0,
                          ngrid=40,
                          rho_thresh=1e-3):
    xs = np.linspace(-box, box, ngrid)
    ys = np.linspace(-box, box, ngrid)
    zs = np.linspace(-box, box, ngrid)
    X, Y, Z = np.meshgrid(xs, ys, zs, indexing="ij")
    coords = np.stack([X.ravel(), Y.ravel(), Z.ravel()], axis=1)

    mf = dft.RKS(mol)
    mf.xc = "PBE"
    mf.max_cycle = 200
    mf.conv_tol = 1e-6
    mf.kernel()

    ni = numint.NumInt()
    ao = ni.eval_ao(mol, coords)
    dm = mf.make_rdm1()
    rho_flat = ni.eval_rho(mol, ao, dm)
    rho = rho_flat.reshape((ngrid, ngrid, ngrid))

    dx = xs[1] - xs[0]
    lap = (
        np.gradient(np.gradient(rho, dx, axis=0), dx, axis=0) +
        np.gradient(np.gradient(rho, dx, axis=1), dx, axis=1) +
        np.gradient(np.gradient(rho, dx, axis=2), dx, axis=2)
    )

    mask = rho > rho_thresh
    kappa = np.abs(lap[mask].ravel())
    if kappa.size == 0:
        print("Warning: no points above rho_thresh in grid")
        return np.array([]), mf
    kappa = np.clip(kappa, 0, np.percentile(kappa, 99.5))
    return kappa, mf

def cluster_homo_cloud(mol, mf,
                       box=4.0,
                       ngrid=40,
                       keep_percent=97.0,
                       max_points=5000):
    xs = np.linspace(-box, box, ngrid)
    ys = np.linspace(-box, box, ngrid)
    zs = np.linspace(-box, box, ngrid)
    X, Y, Z = np.meshgrid(xs, ys, zs, indexing="ij")
    coords = np.stack([X.ravel(), Y.ravel(), Z.ravel()], axis=1)

    ni = numint.NumInt()
    ao = ni.eval_ao(mol, coords)

    mo_coeff = mf.mo_coeff   # shape (nao, nmo) for RKS
    mo_occ   = mf.mo_occ     # shape (nmo,)
    occ_idx  = np.where(mo_occ > 1e-6)[0]
    homo_idx = occ_idx[-1]

    mo_vals = ao @ mo_coeff[:, homo_idx]
    mo_vals = np.real(mo_vals)
    prob = mo_vals**2

    thr = np.percentile(prob, keep_percent)
    mask = prob > thr
    coords_cloud = coords[mask]
    psi_cloud    = mo_vals[mask]

    if coords_cloud.shape[0] > max_points:
        idx = np.random.choice(coords_cloud.shape[0], max_points, replace=False)
        coords_cloud = coords_cloud[idx]
        psi_cloud    = psi_cloud[idx]

    return coords_cloud, psi_cloud

# =========================================
# 5. Plotting: geometry + curvature PDF + HOMO cloud
# =========================================

def plot_cluster_fig(symbols, coords_atoms, kappa_fes, coords_cloud, psi_cloud, cid):
    fig = plt.figure(figsize=(15, 4))

    # (a) geometry
    ax1 = fig.add_subplot(1, 3, 1, projection='3d')
    color_atoms = {"Fe": "red", "S": "yellow"}
    for (x, y, z), sym in zip(coords_atoms, symbols):
        ax1.scatter(x, y, z, s=80,
                    color=color_atoms.get(sym, "grey"),
                    edgecolors="k")
        ax1.text(x, y, z + 0.3, sym, ha='center', fontsize=9)
    for i in range(len(coords_atoms)):
        for j in range(i+1, len(coords_atoms)):
            d = np.linalg.norm(coords_atoms[i] - coords_atoms[j])
            if d < 3.0:
                ax1.plot([coords_atoms[i,0], coords_atoms[j,0]],
                         [coords_atoms[i,1], coords_atoms[j,1]],
                         [coords_atoms[i,2], coords_atoms[j,2]],
                         color="k", linewidth=0.7)
    ax1.set_title(f"Fe–S cluster from 6LK1 (cluster {cid})")
    ax1.set_xticks([]); ax1.set_yticks([]); ax1.set_zticks([])
    max_range = (coords_atoms.max(axis=0) - coords_atoms.min(axis=0)).max()
    mid = coords_atoms.mean(axis=0)
    for setter, m in zip([ax1.set_xlim, ax1.set_ylim, ax1.set_zlim], mid):
        setter(m - max_range/2, m + max_range/2)

    # (b) curvature PDF
    ax2 = fig.add_subplot(1, 3, 2)
    if kappa_fes.size > 0:
        counts, edges = np.histogram(kappa_fes, bins=60, density=True)
        eps = 1e-12
        counts = np.maximum(counts, eps)
        centers = 0.5*(edges[:-1] + edges[1:])
        ax2.bar(centers, counts,
                width=(edges[1]-edges[0]),
                color="orange", alpha=0.9)
        ax2.set_yscale("log")
    ax2.set_xlabel(r"Electron curvature $|\nabla^2 \rho|$ (arb.)")
    ax2.set_ylabel("Probability density (log scale)")
    ax2.set_title("Fe–S curvature PDF")

    # (c) HOMO cloud
    ax3 = fig.add_subplot(1, 3, 3, projection='3d')
    colors_cloud = np.where(psi_cloud >= 0, "blue", "red")
    ax3.scatter(coords_cloud[:,0],
                coords_cloud[:,1],
                coords_cloud[:,2],
                s=3, c=colors_cloud, alpha=0.6)
    for (x, y, z), sym in zip(coords_atoms, symbols):
        ax3.scatter(x, y, z, s=60,
                    color=color_atoms.get(sym, "grey"),
                    edgecolors="k")
    ax3.set_title("Fe–S HOMO orbital cloud")
    ax3.set_xticks([]); ax3.set_yticks([]); ax3.set_zticks([])
    for setter, m in zip([ax3.set_xlim, ax3.set_ylim, ax3.set_zlim], mid):
        setter(m - max_range/2, m + max_range/2)

    plt.tight_layout()
    outname = f"fes_6lk1_cluster{cid}.png"
    plt.savefig(outname, dpi=300)
    plt.show()
    print("Saved", outname)

# =========================================
# 6. Run for both 2Fe–6S clusters (1 and 5)
# =========================================

for cid in [1, 5]:
    print("\n========== CLUSTER", cid, "==========")
    mol, sym_array, coords_centered = build_cluster_mol(cid)
    kappa_fes, mf = cluster_curvature_pdf(mol)
    coords_cloud, psi_cloud = cluster_homo_cloud(mol, mf)
    plot_cluster_fig(sym_array, coords_centered, kappa_fes, coords_cloud, psi_cloud, cid)
```

```python
# =========================================
# Fe–S cluster from 6LK1:
#   - pick a 2Fe–6S cluster (cluster 1 by default)
#   - compute electron curvature PDF |∇²ρ|
#   - apply "breathing" distortions (scale coords)
#   - overlay curvature PDFs for different amplitudes
#   - show HOMO cloud for base vs strongest distortion
# =========================================

import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401

import gemmi
from pyscf import gto, dft
from pyscf.dft import numint

# ---------- 1. Load 6LK1 and collect Fe/S atoms ----------

cif_path = "6LK1.cif"  # adjust if needed

doc = gemmi.cif.read_file(cif_path)
block = doc.sole_block()
structure = gemmi.make_structure_from_block(block)
model = structure[0]

fes_atoms_all = []
for chain in model:
    for res in chain:
        for atom in res:
            el = atom.element.name
            if el.upper().startswith("FE") or el.upper() == "S":
                fes_atoms_all.append({
                    "element": "Fe" if el.upper().startswith("FE") else "S",
                    "x": atom.pos.x,
                    "y": atom.pos.y,
                    "z": atom.pos.z,
                })

fes_coords = np.array([[a["x"], a["y"], a["z"]] for a in fes_atoms_all])
fes_elems  = np.array([a["element"] for a in fes_atoms_all])

print(f"Found {len(fes_atoms_all)} Fe/S atoms in 6LK1")

# ---------- 2. Cluster Fe/S atoms (distance-based) ----------

def cluster_fes(coords, cutoff=3.0):
    N = coords.shape[0]
    adj = [[] for _ in range(N)]
    for i in range(N):
        for j in range(i+1, N):
            d = np.linalg.norm(coords[i] - coords[j])
            if d < cutoff:
                adj[i].append(j)
                adj[j].append(i)
    visited = [False]*N
    clusters = []
    for i in range(N):
        if not visited[i]:
            stack = [i]
            comp = []
            visited[i] = True
            while stack:
                k = stack.pop()
                comp.append(k)
                for nb in adj[k]:
                    if not visited[nb]:
                        visited[nb] = True
                        stack.append(nb)
            clusters.append(comp)
    return clusters

clusters = cluster_fes(fes_coords, cutoff=3.0)
print(f"Found {len(clusters)} Fe/S-connected clusters")
for idx, comp in enumerate(clusters):
    n_fe = np.sum(fes_elems[comp] == "Fe")
    n_s  = np.sum(fes_elems[comp] == "S")
    print(f"  Cluster {idx}: {len(comp)} atoms ({n_fe} Fe, {n_s} S)")

# ---------- 3. Build centred PySCF molecule for chosen Fe–S cluster ----------





def build_cluster_mol(cluster_idx):
    atom_indices = clusters[cluster_idx]
    print("\nUsing cluster", cluster_idx, "with atoms:")
    cluster_coords = fes_coords[atom_indices]
    center = cluster_coords.mean(axis=0)

    atom_lines = []
    for i in atom_indices:
        sym = fes_elems[i]
        x, y, z = fes_coords[i] - center
        atom_lines.append(f"{sym} {x:.6f} {y:.6f} {z:.6f}")
        # FIXED THIS LINE ↓↓↓
        print(f"  {i}: {sym}, ({fes_coords[i,0]:.3f}, {fes_coords[i,1]:.3f}, {fes_coords[i,2]:.3f})")

    atom_str = "\n".join(atom_lines)
    print("\nPySCF atom block (centred):\n", atom_str)

    mol = gto.Mole()
    mol.atom = atom_str
    mol.basis = "sto-3g"
    mol.charge = 0
    mol.spin = 0
    mol.build()

    symbols = np.array([line.split()[0] for line in atom_lines])
    coords_centered = cluster_coords - center
    return mol, symbols, coords_centered
# Pick the first 2Fe–6S cluster (cluster 1). Change to 5 if desired.
cluster_idx = 5
mol_base, symbols, coords_base = build_cluster_mol(cluster_idx)

# ---------- 4. Curvature PDF for a given set of coords ----------

def curvature_pdf_from_coords(symbols, coords,
                              box=4.0,
                              ngrid=40,
                              rho_thresh=1e-3):
    xs = np.linspace(-box, box, ngrid)
    ys = np.linspace(-box, box, ngrid)
    zs = np.linspace(-box, box, ngrid)
    X, Y, Z = np.meshgrid(xs, ys, zs, indexing="ij")
    coords_grid = np.stack([X.ravel(), Y.ravel(), Z.ravel()], axis=1)

    # Build molecule for these coords
    atom_lines = [f"{s} {x:.6f} {y:.6f} {z:.6f}"
                  for (s, (x, y, z)) in zip(symbols, coords)]
    atom_str = "\n".join(atom_lines)

    mol = gto.Mole()
    mol.atom = atom_str
    mol.basis = "sto-3g"
    mol.charge = 0
    mol.spin = 0
    mol.build()

    # RKS PBE
    mf = dft.RKS(mol)
    mf.xc = "PBE"
    mf.max_cycle = 200
    mf.conv_tol = 1e-6
    mf.kernel()

    ni = numint.NumInt()
    ao = ni.eval_ao(mol, coords_grid)
    dm = mf.make_rdm1()
    rho_flat = ni.eval_rho(mol, ao, dm)
    rho = rho_flat.reshape((ngrid, ngrid, ngrid))

    dx = xs[1] - xs[0]
    lap = (
        np.gradient(np.gradient(rho, dx, axis=0), dx, axis=0) +
        np.gradient(np.gradient(rho, dx, axis=1), dx, axis=1) +
        np.gradient(np.gradient(rho, dx, axis=2), dx, axis=2)
    )

    mask = rho > rho_thresh
    kappa = np.abs(lap[mask].ravel())
    if kappa.size == 0:
        print("Warning: no points above rho_thresh")
        return np.array([]), mf, mol
    kappa = np.clip(kappa, 0, np.percentile(kappa, 99.5))
    return kappa, mf, mol

# ---------- 5. HOMO cloud for given coords/mf ----------

def homo_cloud_from_mf(mol, mf,
                       box=4.0,
                       ngrid=40,
                       keep_percent=97.0,
                       max_points=5000):
    xs = np.linspace(-box, box, ngrid)
    ys = np.linspace(-box, box, ngrid)
    zs = np.linspace(-box, box, ngrid)
    X, Y, Z = np.meshgrid(xs, ys, zs, indexing="ij")
    coords = np.stack([X.ravel(), Y.ravel(), Z.ravel()], axis=1)

    ni = numint.NumInt()
    ao = ni.eval_ao(mol, coords)

    mo_coeff = mf.mo_coeff
    mo_occ   = mf.mo_occ
    occ_idx  = np.where(mo_occ > 1e-6)[0]
    homo_idx = occ_idx[-1]

    mo_vals = ao @ mo_coeff[:, homo_idx]
    mo_vals = np.real(mo_vals)
    prob = mo_vals**2

    thr = np.percentile(prob, keep_percent)
    mask = prob > thr
    coords_cloud = coords[mask]
    psi_cloud    = mo_vals[mask]

    if coords_cloud.shape[0] > max_points:
        idx = np.random.choice(coords_cloud.shape[0], max_points, replace=False)
        coords_cloud = coords_cloud[idx]
        psi_cloud    = psi_cloud[idx]

    return coords_cloud, psi_cloud

# ---------- 6. Apply "breathing" distortions and compute PDFs ----------

# scale factors: 0 = base, >0 = uniform expansion (vibrational amplitude)
scale_factors = [0.0, 0.05, 0.10, 0.15]

kappa_by_scale = {}
mf_by_scale = {}
mol_by_scale = {}

for s in scale_factors:
    coords_scaled = coords_base * (1.0 + s)
    print(f"\n=== Scale factor {s:.2f} ===")
    kappa_s, mf_s, mol_s = curvature_pdf_from_coords(symbols, coords_scaled)
    kappa_by_scale[s] = kappa_s
    mf_by_scale[s]    = mf_s
    mol_by_scale[s]   = mol_s

# HOMO clouds for base and max scale
base_scale = scale_factors[0]
max_scale  = scale_factors[-1]
coords_cloud_base, psi_cloud_base = homo_cloud_from_mf(mol_by_scale[base_scale],
                                                       mf_by_scale[base_scale])
coords_cloud_max,  psi_cloud_max  = homo_cloud_from_mf(mol_by_scale[max_scale],
                                                       mf_by_scale[max_scale])

# ---------- 7. Plot: geometry (base), PDFs overlay, HOMO clouds (base vs max) ----------

fig = plt.figure(figsize=(15, 4))

# (a) geometry (base)
ax1 = fig.add_subplot(1, 3, 1, projection='3d')
color_atoms = {"Fe": "red", "S": "yellow"}
for (x, y, z), sym in zip(coords_base, symbols):
    ax1.scatter(x, y, z, s=80,
                color=color_atoms.get(sym, "grey"),
                edgecolors="k")
    ax1.text(x, y, z + 0.3, sym, ha='center', fontsize=9)
# bonds
for i in range(len(coords_base)):
    for j in range(i+1, len(coords_base)):
        d = np.linalg.norm(coords_base[i] - coords_base[j])
        if d < 3.0:
            ax1.plot([coords_base[i,0], coords_base[j,0]],
                     [coords_base[i,1], coords_base[j,1]],
                     [coords_base[i,2], coords_base[j,2]],
                     color="k", linewidth=0.7)

ax1.set_title(f"Fe–S cluster (6LK1, cluster {cluster_idx})")
ax1.set_xticks([]); ax1.set_yticks([]); ax1.set_zticks([])
max_range = (coords_base.max(axis=0) - coords_base.min(axis=0)).max()
mid = coords_base.mean(axis=0)
for setter, m in zip([ax1.set_xlim, ax1.set_ylim, ax1.set_zlim], mid):
    setter(m - max_range/2, m + max_range/2)

# (b) curvature PDFs overlay
ax2 = fig.add_subplot(1, 3, 2)
colors = ["tab:blue", "tab:green", "tab:orange", "tab:red"]
for s, c in zip(scale_factors, colors):
    kappa = kappa_by_scale[s]
    if kappa.size == 0:
        continue
    counts, edges = np.histogram(kappa, bins=60, density=True)
    eps = 1e-12
    counts = np.maximum(counts, eps)
    centers = 0.5 * (edges[:-1] + edges[1:])
    ax2.plot(centers, counts, color=c, label=f"scale={s:.2f}")
ax2.set_yscale("log")
ax2.set_xlabel(r"Electron curvature $|\nabla^2 \rho|$ (arb.)")
ax2.set_ylabel("Probability density (log scale)")
ax2.set_title("Fe–S curvature PDFs under 'breathing' distortion")
ax2.legend()

# (c) HOMO clouds: base vs max scale
ax3 = fig.add_subplot(1, 3, 3, projection='3d')
# base (blue/red)
cols_base = np.where(psi_cloud_base >= 0, "blue", "red")
ax3.scatter(coords_cloud_base[:,0],
            coords_cloud_base[:,1],
            coords_cloud_base[:,2],
            s=4, c=cols_base, alpha=0.5, label="base")
# max-distorted (cyan/magenta)
cols_max = np.where(psi_cloud_max >= 0, "cyan", "magenta")
ax3.scatter(coords_cloud_max[:,0],
            coords_cloud_max[:,1],
            coords_cloud_max[:,2],
            s=4, c=cols_max, alpha=0.5, label=f"scale={max_scale:.2f}")
# atoms
for (x, y, z), sym in zip(coords_base, symbols):
    ax3.scatter(x, y, z, s=80,
                color=color_atoms.get(sym, "grey"),
                edgecolors="k")
    ax3.text(x, y, z + 0.3, sym, ha='center', fontsize=8)

ax3.set_title("Fe–S HOMO cloud\n(base vs vibrationally distorted)")
ax3.set_xticks([]); ax3.set_yticks([]); ax3.set_zticks([])
for setter, m in zip([ax3.set_xlim, ax3.set_ylim, ax3.set_zlim], mid):
    setter(m - max_range/2, m + max_range/2)
ax3.legend(loc="upper right")

plt.tight_layout()
plt.savefig("fes_6LK1_vibration_curvature.png", dpi=300)
plt.show()
print("Saved figure: fes_6LK1_vibration_curvature.png")
```
